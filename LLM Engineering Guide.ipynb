## 1. AI, ML, DL, NLP & LLMs

### Understanding the Key Concepts

AI, ML, DL, NLP, and LLMs are closely related concepts, but **they do not all describe the same kind of thing**. Some describe a broad research field, some describe learning methods, and others describe tasks or model families.

| Concept | Core idea | Typical examples |
|---|---|---|
| **Artificial Intelligence (AI)** | The broad goal of making systems perform tasks associated with perception, prediction, planning, or language | Planning systems, expert systems, autonomous agents |
| **Machine Learning (ML)** | Learns patterns from data rather than relying only on hand-written rules | Classification, regression, recommendation |
| **Deep Learning (DL)** | Uses multi-layer neural networks to learn representations | CNNs, RNNs, Transformers |
| **Natural Language Processing (NLP)** | Concerns computational processing of human language | Classification, parsing, translation, retrieval, generation |
| **Language Model (LM)** | Assigns probabilities to token sequences or predicts tokens in context | N-gram models, neural language models |
| **Large Language Model (LLM)** | A language model scaled in data, parameters, and compute to support a broad range of language tasks | GPT-style language models |
| **Generative AI** | A broad category of systems that create content | Text-generation models, image diffusion models |
| **Foundation Model** | A broadly pretrained model that can be adapted to multiple downstream tasks | Language, vision, and multimodal foundation models |

**Important distinction:** These concepts overlap rather than forming one strict hierarchy.

- AI includes methods that do not use machine learning.
- NLP overlaps with ML and DL, but NLP is a task-oriented field rather than a single learning algorithm.
- LLMs are language models, but not every language model is large.
- Foundation models are broader than text-only LLMs.
- Generative AI includes models beyond LLMs, such as image diffusion models.

### How a Language Model Learns to Predict Text?

Text is first divided into **tokens**. A language model learns statistical patterns that help it assign probabilities to sequences or predict the next token from the preceding context.

For an autoregressive language model, the central task is:

$$
\boxed{
P(x_t \mid x_1, x_2, \ldots, x_{t-1})
}
$$

Here:

- $x_t$ is the token being predicted.
- $x_1,\ldots,x_{t-1}$ are the preceding tokens.
- $P(x_t \mid x_1,\ldots,x_{t-1})$ is the probability assigned to the next token given its context.

For example, after receiving the text *"The sun rises in the"*, a language model may assign probabilities to several possible next tokens. It predicts a distribution rather than simply looking up one universally correct continuation.

For a complete sequence, the autoregressive probability factorization is:

$$
\boxed{
P(x_1,\ldots,x_T)
=
\prod_{t=1}^{T}
P(x_t \mid x_1,\ldots,x_{t-1})
}
$$

This factorization is the mathematical foundation of next-token language modeling. It does not assume that tokens are independent; each prediction can depend on the available preceding context.

````python
import sys, platform, math, re, random, time, json, textwrap
from collections import Counter, defaultdict
from typing import Iterable, Sequence

import numpy as np
import matplotlib.pyplot as plt

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

print(f"Python: {platform.python_version()} | NumPy: {np.__version__}")
print("Random seed:", SEED)
````

````python
# A runnable miniature ML experiment: rule-based vs. learned text classification.
# This small Multinomial Naive Bayes model learns word statistics from examples;
# it is a teaching experiment, not a general-purpose sentiment system.

def tokenize_toy(text: str) -> list[str]:
    return re.findall(r"[a-z]+", text.lower())

train_examples = [
    ("positive", "good enjoyable movie"),
    ("positive", "great bright story"),
    ("positive", "love this excellent film"),
    ("positive", "happy fun wonderful"),
    ("negative", "bad boring movie"),
    ("negative", "awful dark story"),
    ("negative", "hate this terrible film"),
    ("negative", "sad dull disappointing"),
]
test_examples = [
    ("positive", "good excellent film"),
    ("negative", "terrible boring story"),
    ("positive", "wonderful fun movie"),
    ("negative", "dark dull film"),
]

classes = sorted({label for label, _ in train_examples})
class_document_counts = Counter(label for label, _ in train_examples)
class_word_counts = {label: Counter() for label in classes}
class_token_totals = Counter()
vocabulary = set()
for label, text in train_examples:
    words = tokenize_toy(text)
    vocabulary.update(words)
    class_word_counts[label].update(words)
    class_token_totals[label] += len(words)
vocabulary = sorted(vocabulary)
alpha = 1.0  # Laplace smoothing for word likelihoods.

log_prior = {
    label: math.log(class_document_counts[label] / len(train_examples))
    for label in classes
}
log_likelihood = {
    label: {
        word: math.log((class_word_counts[label][word] + alpha) /
                       (class_token_totals[label] + alpha * len(vocabulary)))
        for word in vocabulary
    }
    for label in classes
}

def naive_bayes_scores(text: str) -> dict[str, float]:
    words = tokenize_toy(text)
    return {
        label: log_prior[label] + sum(log_likelihood[label].get(word, 0.0) for word in words)
        for label in classes
    }

def naive_bayes_predict(text: str) -> str:
    return max(naive_bayes_scores(text), key=naive_bayes_scores(text).get)

positive_words = {"good", "great", "love", "excellent", "fun", "wonderful", "happy"}
negative_words = {"bad", "hate", "boring", "awful", "terrible", "dull", "sad", "dark"}
def rule_based_predict(text: str) -> str:
    words = set(tokenize_toy(text))
    positive_hits = len(words & positive_words)
    negative_hits = len(words & negative_words)
    if positive_hits == negative_hits:
        return "uncertain"
    return "positive" if positive_hits > negative_hits else "negative"

print(f"Training examples: {len(train_examples)} | vocabulary size: {len(vocabulary)}")
print("The naive Bayes model learns class priors and smoothed word likelihoods from the training examples.\n")
correct = 0
for expected, text in test_examples:
    rule_pred = rule_based_predict(text)
    model_scores = naive_bayes_scores(text)
    model_pred = max(model_scores, key=model_scores.get)
    correct += model_pred == expected
    score_text = ", ".join(f"{k}={v:.2f}" for k, v in model_scores.items())
    print(f"{text!r:32} expected={expected:8} rule={rule_pred:9} naive_bayes={model_pred:8} | log-scores: {score_text}")
print(f"Held-out accuracy on this tiny illustrative set: {correct}/{len(test_examples)} = {correct/len(test_examples):.1%}")
print("Do not interpret this tiny hand-built result as a real-world benchmark.")
````

### A Short History of Language Modeling

Language modeling has evolved through several major approaches. Each family introduced different ways to represent context and learn patterns from language.

| Stage | Model family | Main idea | Important limitation or change |
|---|---|---|---|
| **1** | Rule-based systems | Use explicit grammars, lexicons, and templates | Interpretable, but costly to extend to exceptions and varied language |
| **2** | Statistical language models | Estimate sequence probabilities from n-gram counts, often with smoothing | Context is limited to a small number of recent tokens |
| **3** | Distributed representations and neural LMs | Use embeddings and neural layers to generalize across similar contexts | Require learned parameters and appropriate training data |
| **4** | RNNs, LSTMs, and sequence-to-sequence models | Maintain recurrent state while processing sequences | Training can be sequential, and long-range dependencies can be difficult |
| **5** | Transformers | Use self-attention for contextual information exchange and parallelizable training | Standard dense self-attention has quadratic pairwise interactions in sequence length |
| **6** | Instruction-tuned and tool-using systems | Adapt pretrained models to follow instructions, retrieve external information, or call constrained tools | Require evaluation, security controls, and reliable tool integration |
| **7** | Multimodal and alternative sequence architectures | Combine text with image, audio, or video representations, or explore state-space and hybrid backbones | Performance and capabilities depend on architecture, data, implementation, and task |

The Transformer architecture was introduced in the paper by Vaswani et al. (2017), *Attention Is All You Need*.

### Different Model Families Solve Different Problems

Not every model is designed to generate text in the same way. Three important Transformer families are distinguished by how they process input and produce output.

| Architecture | How it processes information | Typical applications |
|---|---|---|
| **Encoder-only** | Builds bidirectional contextual representations | Classification, embeddings, token labeling |
| **Decoder-only** | Predicts continuations using preceding tokens | Text generation, chat, code completion |
| **Encoder–decoder** | Encodes an input sequence and generates an output sequence | Translation, summarization, conditional text transformation |

Examples include BERT-style encoder-only models, GPT-style decoder-only models, and T5-style encoder–decoder models.

**Diffusion models are different:** they generally generate content through iterative denoising rather than autoregressive next-token prediction. They are commonly used for image generation and other generative tasks.

### Key Takeaway

AI is the broad field; ML and DL describe learning approaches; NLP focuses on computational language tasks; and language models learn probability distributions over tokens or sequences. LLMs and foundation models extend these ideas through large-scale pretraining, while different architectures support different objectives.

Understanding these distinctions provides the foundation for studying tokenization, neural networks, language modeling, attention, Transformers, fine-tuning, and modern LLM applications.

## 2. Mathematical Foundations

### 2.1. Scalars, Vectors, Matrices, and Tensors

Mathematics provides the language used to represent data, model parameters, and computations inside neural networks.

#### Understanding the different types of data

| Mathematical object | Number of dimensions (rank) | Example | Shape |
|---|---:|---|---|
| Scalar | 0 | A loss value | `()` |
| Vector | 1 | A feature vector | `(3,)` |
| Matrix | 2 | A table of features | `(2, 3)` |
| Tensor | 3 or more | A batch of token representations | `(B, T, C)` |

A **tensor** is a general mathematical representation of numerical data. Its shape describes how its elements are organized.

For example, a batch of token representations is commonly represented as:

$$
\boxed{X \in \mathbb{R}^{B \times T \times C}}
$$

where:

- $X$ is the input tensor.
- $B$ is the batch size (number of sequences processed together).
- $T$ is the sequence length (number of token positions).
- $C$ is the number of features or embedding dimensions.
- $\mathbb{R}$ denotes the set of real numbers.

**Example:** Suppose a batch contains 4 sentences, each represented using 10 token positions and 64 features per token.

$$
\boxed{X \in \mathbb{R}^{4 \times 10 \times 64}}
$$

The total number of scalar values is:

$$
\boxed{4 \times 10 \times 64 = 2560}
$$

Each token position has a 64-dimensional representation.

**Why this matters for LLMs:** Tensor shapes determine whether operations such as embedding lookup, attention, matrix multiplication, and batch processing are mathematically compatible.

---

### 2.2. Dot Products & Matrix Multiplication

#### A. Dot Product

The dot product combines two vectors into a single number by multiplying corresponding elements and adding the results.

For two vectors $x,y \in \mathbb{R}^{d}$:

$$
\boxed{
x^\top y = \sum_{i=1}^{d} x_i y_i
}
$$

Here:

- $x$ and $y$ are vectors of length $d$.
- $x_i$ and $y_i$ are their respective elements at position $i$.
- $x^\top$ denotes the transpose of $x$.
- $\sum$ means to add the products across all positions.

**Example**

Let:

$$
\boxed{x =
\begin{bmatrix}
1\\
2\\
3
\end{bmatrix},
\qquad
y =
\begin{bmatrix}
4\\
5\\
6
\end{bmatrix}}
$$

Their dot product is:

$$
\boxed{\begin{aligned}
x^\top y
&= (1)(4)+(2)(5)+(3)(6)\\
&=4+10+18\\
&=\boxed{32}
\end{aligned}}
$$

The result is a scalar, not another vector.

**Connection to Attention:** Transformer attention uses dot products between query and key vectors to calculate compatibility scores. These scores help determine how information from different token positions is combined.

#### B. Matrix Multiplication

Consider two matrices:

$$
\boxed{A \in \mathbb{R}^{m \times n},
\qquad
B \in \mathbb{R}^{n \times p}}
$$

Their product is:

$$
\boxed{
C=AB \in \mathbb{R}^{m \times p}
}
$$

The rule is simple:

**The inner dimensions must match. The outer dimensions determine the output shape.**

$$
\boxed{\underbrace{(m \times n)}_{A}
\;\times\;
\underbrace{(n \times p)}_{B}
\;=\;
\underbrace{(m \times p)}_{C}}
$$

**Example:**

$$
\boxed{\begin{bmatrix}
1 & 2\\
3 & 4
\end{bmatrix}
\begin{bmatrix}
5 & 6\\
7 & 8
\end{bmatrix}
=
\begin{bmatrix}
19 & 22\\
43 & 50
\end{bmatrix}}
$$

The top-left result is:

$$
\boxed{(1)(5)+(2)(7)=19}
$$

Each output element is obtained by taking a row from the first matrix and a column from the second matrix, then computing their dot product.

**Why it matters:** Neural network layers commonly transform their input using a matrix multiplication followed by a bias addition and an activation function.

### 2.3. Probability Distributions

A probability distribution assigns a probability to each possible outcome.

For a categorical distribution with $K$ possible outcomes:

$$
\boxed{p_i \geq 0
\qquad \text{for every } i}
$$

and

$$
\boxed{
\sum_{i=1}^{K} p_i=1
}
$$

The first condition says that probabilities cannot be negative. The second says that all possible outcomes together have probability 1.

**Example: next-token prediction**

Suppose a language model assigns the following probabilities to three possible next tokens:

| Candidate token | Probability |
|---|---:|
| `cat` | 0.70 |
| `dog` | 0.20 |
| `bird` | 0.10 |
| **Total** | **1.00** |

The model considers `cat` the most likely next token because it has the highest predicted probability.

However, a probability of 0.70 does not guarantee that `cat` is the correct continuation. It represents the model's predicted distribution, not certainty about the answer.

---

### 2.4. Entropy: Measuring Uncertainty

Entropy measures the uncertainty associated with a probability distribution.

For a discrete probability distribution $p$:

$$
\boxed{
H(p)=-\sum_{i=1}^{K}p_i\log p_i
}
$$

where:

- $H(p)$ is the entropy.
- $p_i$ is the probability of outcome $i$.
- $\log$ is the natural logarithm unless another base is explicitly specified.
- $K$ is the number of possible outcomes.

The negative sign ensures that entropy is non-negative because probabilities lie between 0 and 1.

#### Intuitive interpretation

- A distribution concentrated on one outcome has low entropy.
- A distribution spread evenly across many outcomes has high entropy.
- For a fixed number of outcomes, the uniform distribution has maximum entropy.

**Example:** Consider two distributions over two outcomes:

$$
\boxed{p=\begin{bmatrix}0.99 & 0.01\end{bmatrix}}
$$

$$
\boxed{q=\begin{bmatrix}0.50 & 0.50\end{bmatrix}}
$$

For $p$, the model strongly favors one outcome. For $q$, both outcomes are equally likely.

Using natural logarithms, their entropies are approximately:

$$
\boxed{H(p)\approx 0.056 \text{ nats}}
$$

$$
\boxed{H(q)\approx 0.693 \text{ nats}}
$$

Therefore:

$$
\boxed{H(q)>H(p)}
$$

The second distribution expresses greater uncertainty.

---

### 2.5. Cross-Entropy: Measuring Prediction Error

Cross-entropy measures how well a predicted probability distribution $p$ matches a target distribution $y$.

It is defined as:

$$
\boxed{
H(y,p)=-\sum_{i=1}^{K}y_i\log p_i
}
$$

where:

- $y_i$ is the target probability for class $i$.
- $p_i$ is the model's predicted probability for class $i$.
- $K$ is the number of possible classes.

When $y$ is a one-hot distribution, the target probability is 1 for the correct class and 0 for all other classes.

If the correct class has index $c$, then:

$$
\boxed{y_c=1,
\qquad
y_i=0 \text{ for } i\neq c}
$$

Substituting into the cross-entropy equation gives:

$$
\boxed{
L=-\log p_c
}
$$

This is the negative log-likelihood loss for one categorical prediction.

** For Example:**

Suppose the correct next token is `cat`.

**Case 1: The model assigns the correct token probability 0.90.**

$$
\boxed{L=-\log(0.90)\approx 0.1054}
$$

**Case 2: The model assigns the correct token probability 0.10.**

$$
\boxed{L=-\log(0.10)\approx 2.3026}
$$

Comparison:

| Probability assigned to the correct token | Cross-entropy loss |
|---:|---:|
| 0.90 | 0.1054 |
| 0.50 | 0.6931 |
| 0.10 | 2.3026 |

The lower the probability assigned to the correct token, the larger the loss.

**Connection to LLM training:** During next-token prediction, a model generates logits, converts them into probabilities, and is penalized according to the probability assigned to the actual next token. Training adjusts the model parameters to reduce this loss across many token positions and training examples.

For a sequence with $T$ target positions, a common average token-level objective is:

$$
\boxed{
L_{\text{sequence}}
=
-\frac{1}{T}
\sum_{t=1}^{T}
\log P(x_t\mid x_1,\ldots,x_{t-1})
}
$$

Here, $x_t$ is the true token at position $t$. Padding positions or other excluded targets may be masked out in actual implementations.

### 2.6. Derivatives: Understanding Change

A derivative describes how a function changes when its input changes.

For a differentiable function $f(w)$, its derivative at $w$ is:

$$
\boxed{
f'(w)=\frac{df}{dw}
}
$$

The derivative measures the instantaneous rate of change of $f$ with respect to $w$.

An intuitive way to understand it is through the slope of a tangent line.

**For Example:** Let,

$$
\boxed{f(w)=w^2}
$$

Differentiating gives:

$$
\boxed{
\frac{df}{dw}=2w
}
$$

At $w=3$:

$$
\boxed{f'(3)=2(3)=6}
$$

This means the local rate of change of the function at $w=3$ is 6.

For a small change $\Delta w$, the corresponding function change is approximately:

$$
\boxed{\Delta f\approx f'(w)\Delta w}
$$

The approximation becomes more accurate as the change becomes sufficiently small for the local linear approximation to hold.

### 2.7. Gradients & the Chain Rule

Neural networks usually contain many trainable parameters, not just one scalar parameter.

If the loss depends on parameters $\theta_1,\theta_2,\ldots,\theta_n$, the gradient is:

$$
\boxed{
\nabla_{\boldsymbol{\theta}}L
=
\begin{bmatrix}
\frac{\partial L}{\partial \theta_1}\\
\frac{\partial L}{\partial \theta_2}\\
\vdots\\
\frac{\partial L}{\partial \theta_n}
\end{bmatrix}
}
$$

Here:

- $L$ is the loss function.
- $\boldsymbol{\theta}$ is the collection of trainable parameters.
- Each partial derivative measures how the loss changes as one parameter changes while the others are held fixed.

In a neural network, the gradient is used to determine how to adjust parameters to reduce the loss.

#### The chain rule

The chain rule computes derivatives of composed functions.

If:

$$
\boxed{z=f(y),
\qquad
y=g(x)}
$$

then:

$$
\boxed{
\frac{dz}{dx}
=
\frac{dz}{dy}\frac{dy}{dx}
}
$$

In words, the change in $z$ with respect to $x$ is calculated by multiplying the change in $z$ with respect to $y$ by the change in $y$ with respect to $x$.

Neural networks contain many composed operations. Backpropagation applies the chain rule repeatedly to propagate gradients from the loss back through the model's layers.

---

### 2.8. Gradient Descent: Learning by Updating Parameters

Gradient descent is an optimization method that adjusts model parameters in the direction opposite to the loss gradient.

For a scalar parameter $w$, the update rule is:

$$
\boxed{
w_{t+1}
=
w_t-\eta\frac{\partial L}{\partial w_t}
}
$$

where:

- $w_t$ is the parameter value at step $t$.
- $w_{t+1}$ is its updated value.
- $L$ is the loss function.
- $\frac{\partial L}{\partial w_t}$ is the derivative of the loss with respect to the parameter.
- $\eta$ is the learning rate, a positive number controlling the update size.

For a vector of parameters, the corresponding update is:

$$
\boxed{
\boldsymbol{\theta}_{t+1}
=
\boldsymbol{\theta}_t
-
\eta\nabla_{\boldsymbol{\theta}}L
}
$$

**Worked Example:**

Suppose the loss function is:

$$
\boxed{L(w)=w^2}
$$

Its derivative is:

$$
\boxed{\frac{dL}{dw}=2w}
$$

Let the initial parameter and learning rate be:

$$
\boxed{w_0=3,
\qquad
\eta=0.1}
$$

The first gradient is:

$$
\boxed{\frac{dL}{dw}\bigg|_{w=3}=6}
$$

Apply the update:

$$
\boxed{\begin{aligned}
w_1
&=w_0-\eta\frac{dL}{dw}\\
&=3-(0.1)(6)\\
&=\boxed{2.4}
\end{aligned}}
$$

The loss changes from:

$$
\boxed{L(w_0)=3^2=9}
$$

to:

$$
\boxed{L(w_1)=2.4^2=5.76}
$$

The parameter moved closer to the minimum at $w=0$, and the loss decreased.

This is a simple illustration with a one-parameter function. Real neural networks use gradients over many parameters, and decreasing the loss on training data does not guarantee good generalization.



#### What happens when the learning rate changes?

- **Too small:** Training may progress slowly.
- **Appropriate:** Updates can reduce the loss steadily.
- **Too large:** Updates may overshoot a minimum or cause unstable training.

The best learning rate depends on the model, data, optimizer, and training configuration.

###  Conditional Probability & Bayes' Theorem

#### 1. Conditional Probability: Probability Given Evidence

Conditional probability measures the probability of an event occurring when we already know that another event has occurred.

It is defined as:

$$
\boxed{
P(A \mid B)=\frac{P(A\cap B)}{P(B)}
}
\qquad \text{where }P(B)>0
$$

**Meaning of each symbol:**

- $P(A \mid B)$: Probability of event $A$, given that event $B$ has occurred.
- $P(A \cap B)$: Probability that both $A$ and $B$ occur.
- $P(B)$: Probability that event $B$ occurs.
- $P(B) > 0$: Ensures that the denominator is nonzero.

**Intuition:** Once we know that $(B)$ has occurred, we restrict our attention to outcomes in which $(B)$ is true. Conditional probability measures how much of that restricted set also satisfies $(A)$.

#### 2. Bayes' Theorem: Updating Probability with Evidence

Bayes' theorem expresses conditional probability in terms of the reverse conditional probability:

$$
\boxed{
P(A\mid B)=\frac{P(B\mid A)\,P(A)}{P(B)}
}
$$

This equation allows us to update our belief about event $(A)$ after observing evidence $(B)$.

**Interpretation of the terms:**

| Term | Meaning |
|---|---|
| $P(A)$ | **Prior:** Probability of $(A)$ before observing the evidence. |
| $P(B \mid A)$ | **Likelihood:** Probability of observing $(B)$ when $(A)$ is true. |
| $P(B)$ | **Evidence:** Overall probability of observing $(B)$. |
| $P(A \mid B)$ | **Posterior:** Updated probability of $(A)$ after observing $()$. |

The relationship can be summarized as:

$$
\boxed{
\text{Posterior}
=
\frac{\text{Likelihood}\times\text{Prior}}
{\text{Evidence}}
}
$$

#### 3. Worked Numerical Example

Suppose:

- $(A)$: It is raining.
- $(B)$: The sky is cloudy.

Assume the following illustrative probabilities:

$$
\boxed{P(A)=0.20}
$$

$$
\boxed{P(B\mid A)=0.80}
$$

$$
\boxed{P(B\mid \neg A)=0.30}
$$

Here, $(-A)$ means that it is not raining.

First, calculate the overall probability of cloudy weather using the law of total probability:

$$
\boxed{\begin{aligned}
P(B)
&=P(B\mid A)P(A)\\
&\quad+P(B\mid\neg A)P(\neg A)\\[4pt]
&=(0.80)(0.20)+(0.30)(0.80)\\[4pt]
&=0.16+0.24\\[4pt]
&=\boxed{0.40}
\end{aligned}}
$$

Now apply Bayes' theorem:

$$
\boxed{\begin{aligned}
P(A\mid B)
&=\frac{P(B\mid A)P(A)}{P(B)}\\[5pt]
&=\frac{(0.80)(0.20)}{0.40}\\[5pt]
&=\frac{0.16}{0.40}\\[5pt]
&=\boxed{0.40}
\end{aligned}}
$$

**Interpretation:** The prior probability of rain was \(20\%\). After observing cloudy weather, the updated probability becomes \(40\%\), under the assumed probabilities.

This illustrates how evidence can change a probability estimate.


#### 4. Connection to Language Models

Language models repeatedly use **conditional probability** to predict tokens from their preceding context.

For a token sequence,

$$
\boxed{x_1,x_2,\ldots,x_{t-1},x_t}
$$

the model predicts the next token using:

$$
\boxed{
P(x_t\mid x_{<t})
}
$$

where:

- $x_t$: The token being predicted.
- $x_{<t}$: Denotes all preceding tokens, namely, $x_1, \ldots, x_{t-1}$.
- $P(x_t \mid x_{<t})$: The model's predicted probability of the next token, given the preceding context.

**Example:** Given the context:

> The cat sat on the ...

A language model might assign probabilities to candidates such as `mat`, `floor`, or `chair`. These probabilities depend on the preceding tokens and the model's learned parameters.

For a complete sequence, the chain rule of probability gives:

$$
\boxed{
P(x_1,x_2,\ldots,x_T)
=
\prod_{t=1}^{T}
P(x_t\mid x_{<t})
}
$$

The first token is handled using the appropriate initial or start-of-sequence distribution.

This factorization explains how an autoregressive language model represents the probability of a complete token sequence through successive next-token probabilities.

**Important distinction:** Using conditional probabilities for next-token prediction is not the same as applying Bayes' theorem to perform explicit inference over every hidden variable in a model. Conditional probability describes how predictions depend on context; Bayes' theorem provides a rule for updating probabilities when evidence is observed.

**Key takeaway:** Conditional probability describes the probability of an event given known information. Bayes' theorem shows how to update a probability using evidence. Language models rely heavily on conditional probabilities to predict the next token and generate sequences of text.

### A differentiable scalar example
Let $L(w) = (w - 3)^2$. Then the derivative is:

$$
\boxed{\frac{dL}{dw} = 2(w - 3)}
$$

If $w = 0$ and $\eta = 0.1$, one gradient descent step updates the parameter as follows:

$$
\boxed{w_{\text{new}} = w - \eta \frac{dL}{dw}}
$$

Substituting the values:

$$
\boxed{w_{\text{new}} = 0 - 0.1 \times 2(0 - 3) = 0.6}
$$

Thus, $w$ moves from $0$ toward the minimum at $w = 3$.

- **Too-large learning rate:** Updates may overshoot the minimum or become unstable.
- **Too-small learning rate:** Training progresses slowly.
- **Modern optimizers:** Methods such as AdamW use adaptive learning-rate scaling and weight decay, but the core learning signal remains the gradient.


````python
w = 0.0
learning_rate = 0.1
history = []
for step in range(25):
    current_loss = (w - 3.0) ** 2
    gradient = 2.0 * (w - 3.0)
    history.append((step, w, current_loss, gradient))
    w -= learning_rate * gradient
print(f"After 25 steps: w={w:.4f}; loss={(w-3.0)**2:.6f}")
arr = np.array(history)
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(arr[:,0], arr[:,2], marker='o', markersize=3)
ax.set_title('Gradient descent on a one-dimensional quadratic')
ax.set_xlabel('Update step'); ax.set_ylabel('Loss'); ax.grid(True, alpha=.3)
plt.show()
````

**Common Mistakes...**

- Comparing logits as if they were probabilities: logits may be negative and need not sum to one.
- Taking `log(0)`: use numerically stable library losses and log-softmax implementations.
- Confusing a transpose with an inverse: they are different operations.
- Treating correlation or similarity as causation.
- Thinking that a low training loss alone proves generalization.

**Recap:** linear algebra moves representations; probability expresses uncertainty; cross-entropy measures predictive mismatch; gradients indicate how parameter changes affect a loss.

## 3. Neural-Network Fundamentals

### From a neuron to a trainable function

A dense layer computes:

$$
\boxed{h = \phi(Wx + b)}
$$

where:

- $x$: Input vector.
- $W$: Weight matrix containing trainable parameters.
- $b$: Bias vector containing trainable parameters.
- $\phi$: Activation function, such as ReLU or GELU.
- $h$: Output (or hidden representation) produced by the layer.

Stacking multiple layers creates a parameterized function. During training:

1. **Forward pass:** Computes predictions from the input.
2. **Loss calculation:** Measures the difference between predictions and target values.
3. **Backpropagation:** Computes gradients of the loss with respect to the model parameters.
4. **Parameter update:** The optimizer uses these gradients to update the parameters and reduce the loss.

### Training, Validation, & Test Sets

These datasets serve different purposes:

- **Train:** Used to estimate model parameters.
- **Validation:** Used to select hyperparameters and model checkpoints.
- **Test:** Used to estimate final generalization performance after model-selection decisions are complete.

**Important:** Do not repeatedly tune the model using the test set, as this can lead to overly optimistic performance estimates.

For sequential or time-series data, split the dataset chronologically whenever future information must not leak into the past.


````python
import torch
from torch import nn
from torch.nn import functional as F

torch.manual_seed(SEED)
if torch.cuda.is_available():
    DEVICE = torch.device('cuda')
elif hasattr(torch.backends, 'mps') and torch.backends.mps.is_available():
    DEVICE = torch.device('mps')
else:
    DEVICE = torch.device('cpu')
print("PyTorch:", torch.__version__, "| selected device:", DEVICE)

# Tiny binary classifier on explicitly synthetic data.
X_syn = torch.randn(256, 2)
y_syn = ((X_syn[:, 0] - 0.6 * X_syn[:, 1]) > 0).long()
net = nn.Sequential(nn.Linear(2, 16), nn.ReLU(), nn.Linear(16, 2))
logits_syn = net(X_syn)
loss_syn = F.cross_entropy(logits_syn, y_syn)
loss_syn.backward()
print("Input shape:", tuple(X_syn.shape), "logits shape:", tuple(logits_syn.shape))
print("Initial loss:", round(loss_syn.item(), 4))
print("Gradient present?", all(p.grad is not None for p in net.parameters()))
print("First weight gradient norm:", round(net[0].weight.grad.norm().item(), 5))
````

### What backpropagation does?

Consider a computation graph:

$$
\boxed{L = f\bigl(g(h(x; W_1); W_2)\bigr)}
$$

The **chain rule** breaks complex derivatives into smaller, connected derivatives. For example:

$$
\boxed{\frac{\partial L}{\partial W_1}
=
\frac{\partial L}{\partial h}
\frac{\partial h}{\partial W_1}}
$$

This allows the model to determine how changes in the earlier-layer parameters $W_1$ affect the final loss $L$.

During backpropagation:

- **Autograd:** Records mathematical operations during the forward pass and calculates derivatives in reverse order.
- **`.backward()`:** Computes gradients and accumulates them in the parameters' `.grad` attributes.
- **`.zero_grad()`:** Clears previously accumulated gradients before the next training iteration.
- **`.step()`:** Updates the model parameters using the gradients and the optimizer's update rule.

**Important:** Calling `.backward()` calculates gradients but does not update the parameters. The optimizer's `.step()` method performs the actual parameter update.


````python
optimizer_demo = torch.optim.AdamW(net.parameters(), lr=0.03)
loss_history_nn = []
for epoch in range(80):
    optimizer_demo.zero_grad(set_to_none=True)
    pred = net(X_syn)
    loss = F.cross_entropy(pred, y_syn)
    loss.backward()
    optimizer_demo.step()
    loss_history_nn.append(loss.item())
with torch.no_grad():
    pred_class = net(X_syn).argmax(dim=1)
    train_accuracy = (pred_class == y_syn).float().mean().item()
print(f"Synthetic training accuracy: {train_accuracy:.3f} (not a real-world benchmark)")
plt.figure(figsize=(7,3)); plt.plot(loss_history_nn)
plt.title('Synthetic classifier training loss'); plt.xlabel('Epoch'); plt.ylabel('Cross-entropy'); plt.grid(alpha=.3); plt.show()
````

**Regularization & Failure Modes**

- **Overfitting:** training performance improves while validation performance worsens.
- **Underfitting:** model or training is insufficient to capture the task.
- **Weight decay:** discourages large weights; in AdamW it is decoupled from the adaptive gradient update.
- **Dropout:** randomly masks activations during training and is disabled during evaluation.
- **Gradient clipping:** limits gradient norm to reduce extreme updates; it does not fix every instability.
- **Normalization:** rescales intermediate activations, but placement and implementation matter.

## 4. NLP & Text Representation

### What Happens Before a Model Sees Text?

A language model does not directly process raw text as humans read it. Text must first be converted into a numerical representation that the model can process. A typical text-processing pipeline is:

$$
\boxed{
\begin{gathered}
\text{Raw Text} \\
\downarrow \\
\text{Normalization (Task-Dependent)} \\
\downarrow \\
\text{Tokenization} \\
\downarrow \\
\text{Token IDs} \\
\downarrow \\
\text{Embedding Lookup} \\
\downarrow \\
\text{Numerical Vectors for the Model}
\end{gathered}
}
$$

Each stage has a distinct purpose:

| Stage | What happens | Example |
|---|---|---|
| **Raw text** | The original human-readable input | `"Language models learn patterns."` |
| **Normalization** | Applies selected text transformations, if needed | Standardizing Unicode or whitespace |
| **Tokenization** | Splits the text into units defined by the tokenizer | Words, subwords, or byte-level units |
| **Token IDs** | Maps each token to an integer from the tokenizer's vocabulary | `[1532, 871, 4921, ...]` |
| **Embedding lookup** | Maps each token ID to a learned vector | A sequence of numerical vectors |

The example token IDs above are illustrative; actual IDs depend on the tokenizer.

### Why Do Modern Tokenizers Use Subwords?

A word-level vocabulary can become very large and may not contain every word encountered during inference. Modern tokenizers therefore commonly use **subword or byte-level units** to balance vocabulary size against sequence length and segmentation behavior. For example, a tokenizer may represent an unfamiliar word using smaller known units instead of requiring a separate vocabulary entry for the entire word.

### Common Tokenization Algorithms

| Algorithm | Basic idea | Important characteristic |
|---|---|---|
| **Byte Pair Encoding (BPE)** | Iteratively merges frequent adjacent token pairs | Builds larger units from smaller units |
| **WordPiece** | Builds subword vocabularies using a model-specific scoring objective | Commonly associated with BERT-style tokenization |
| **Unigram** | Learns a vocabulary of candidate subword units and selects a suitable segmentation model | Uses a different vocabulary-learning objective from BPE |

These algorithms differ in their vocabulary-learning procedures, but all involve trade-offs between vocabulary size, sequence length, and how text is segmented.

**Key idea:** Tokenization affects the entire model pipeline. Different tokenizers can produce different token IDs and sequence lengths for the same input text.

### Toy Byte Pair Encoding (BPE)

**Byte Pair Encoding (BPE)** demonstrates how a vocabulary can grow by repeatedly merging frequent adjacent units. A simplified toy procedure is:

$$
\boxed{
\begin{gathered}
\text{Start with Small Units} \\
\downarrow \\
\text{Count Adjacent Pairs} \\
\downarrow \\
\text{Select a Frequent Pair} \\
\downarrow \\
\text{Merge the Pair into One Unit} \\
\downarrow \\
\text{Repeat for the Desired Number of Merges}
\end{gathered}
}
$$

Each merge can introduce a larger unit into the vocabulary, allowing frequently occurring sequences to be represented with fewer tokens.

#### A Small Illustrative Example

Suppose the corpus contains the sequence:

`l o w`

If the adjacent pair `l o` is selected for merging, the sequence becomes:

`lo w`

If the pair `lo w` is subsequently merged, the sequence becomes:

`low`

This illustrates the core merging idea. The actual merge order depends on pair frequencies and the tokenizer's implementation.

### Why Is This Only a Toy BPE Implementation?

Real tokenizer training includes additional details, such as:

- **Pre-tokenization:** Determines which text regions can be processed together.
- **Byte-level mapping:** Some BPE tokenizers begin from byte representations rather than ordinary characters.
- **Unicode handling:** Determines how text from different writing systems is represented.
- **Merge rules and tie-breaking:** Affect which pairs are selected when counts are equal or when implementation-specific constraints apply.

**Important:** The toy implementation below demonstrates the *idea* of iterative merging. It is not a drop-in replacement for a production tokenizer and should not be expected to reproduce the output of a pretrained model's tokenizer.

### Key Takeaway

Text representation connects human-readable language to numerical model inputs. Normalization prepares text when appropriate, tokenization divides it into units, token IDs map those units to vocabulary entries, and embeddings convert the IDs into vectors. Understanding this pipeline is essential for interpreting sequence length, context-window usage, and model compatibility.

````python
def toy_bpe(words: list[str], n_merges: int = 5):
    """Illustrative character-level BPE: returns tokens and merge history."""
    pieces = [list(w) + ['</w>'] for w in words]
    merges = []
    for _ in range(n_merges):
        pairs = Counter()
        for seq in pieces:
            pairs.update(zip(seq, seq[1:]))
        if not pairs:
            break
        pair, count = pairs.most_common(1)[0]
        if count < 2:
            break
        merged = ''.join(pair)
        new_pieces = []
        for seq in pieces:
            out, i = [], 0
            while i < len(seq):
                if i < len(seq)-1 and (seq[i], seq[i+1]) == pair:
                    out.append(merged); i += 2
                else:
                    out.append(seq[i]); i += 1
            new_pieces.append(out)
        pieces = new_pieces
        merges.append((pair, merged, count))
    return pieces, merges

bpe_tokens, bpe_merges = toy_bpe(["low", "lower", "lowest", "newer", "newest"], n_merges=6)
print("Toy merges:")
for merge in bpe_merges: print(" ", merge)
print("Resulting token sequences:", bpe_tokens)
````

### One-hot Vectors vs. Embeddings

A **one-hot vector** of vocabulary size $V$ contains one $1$ and $V-1$ zeros. It is sparse and does not encode similarity between tokens by itself.

An **embedding table** maps token IDs to dense, learned vector representations:

$$
\boxed{E \in \mathbb{R}^{V \times C}}
$$

where:

- $V$: Vocabulary size.
- $C$: Embedding dimension.
- $E$: Embedding matrix containing the learned token representations.

For a token with ID $i$, its embedding is the corresponding row of the matrix:

$$
\boxed{E_i \in \mathbb{R}^{C}}
$$

The embedding values are learned during training. Contextual meaning emerges as the token representation is processed by the model.

- **Static embeddings:** A token has the same representation regardless of its surrounding context.
- **Contextual embeddings:** A token's representation depends on the surrounding tokens and the context in which it appears.

**Practical note:** Tokenization affects context-window usage, computational cost, multilingual coverage, and how text, code, and numbers are segmented. Token counts cannot be inferred exactly from word counts.


````python
# Real embedding lookup: compare sparse one-hot vectors with a trainable dense table.
torch.manual_seed(SEED)
toy_tokens = ["cat", "dog", "fish", "bird"]
token_to_id = {token: i for i, token in enumerate(toy_tokens)}
sequence_ids = torch.tensor([token_to_id[t] for t in ["cat", "dog", "fish", "cat"]], dtype=torch.long)
one_hot_vectors = F.one_hot(sequence_ids, num_classes=len(toy_tokens)).float()
embedding_layer = nn.Embedding(num_embeddings=len(toy_tokens), embedding_dim=3)
dense_vectors = embedding_layer(sequence_ids)
print("Token IDs:", sequence_ids.tolist())
print("One-hot shape:", tuple(one_hot_vectors.shape), "| embedding shape:", tuple(dense_vectors.shape))
print("First one-hot vector:", one_hot_vectors[0].tolist())
print("First learned embedding (initialization before training):", dense_vectors[0].detach().round(decimals=3).tolist())
# Demonstrate that lookup is differentiable and updates only rows used in the batch.
embedding_layer.zero_grad(set_to_none=True)
dense_vectors = embedding_layer(sequence_ids)
embedding_loss = dense_vectors.square().mean()
embedding_loss.backward()
print("Embedding lookup loss:", round(embedding_loss.item(), 5))
print("Gradient exists:", embedding_layer.weight.grad is not None,
      "| gradient norm:", round(embedding_layer.weight.grad.norm().item(), 5))

# Compare three tokenization schemes on the same short sentence.
comparison_text = "small models learn useful patterns"
char_count = len(comparison_text)
word_tokens_demo = re.findall(r"\w+|[^\w\s]", comparison_text)
bpe_word_pieces, bpe_history = toy_bpe(comparison_text.split(), n_merges=8)
bpe_count = sum(len(pieces) for pieces in bpe_word_pieces)
print("\nTokenization length comparison:")
print({"characters": char_count, "regex words": len(word_tokens_demo), "toy BPE pieces": bpe_count})
print("Regex tokens:", word_tokens_demo)
print("Toy BPE pieces:", bpe_word_pieces)
````

## 5. Language Modeling

### The Autoregressive Factorization

Using the probability chain rule, a sequence can be factored as:

$$
\boxed{P(x_1, \ldots, x_T)
=
\prod_{t=1}^{T}
P(x_t \mid x_1, \ldots, x_{t-1})}
$$

This is a **factorization, not an independence assumption**. Each conditional probability can depend on the full preceding context available to the model.

For the first token, the condition is an empty context or a special start token, depending on the convention used.

A **causal language model** predicts the next token based on previous tokens. During training, all positions can be supervised in parallel because the complete target sequence is known. However, a **causal mask** prevents each position from accessing future tokens.

### N-grams & Perplexity

An **n-gram model** uses a limited recent context. For example, a trigram model estimates:

$$
\boxed{P(x_t \mid x_{t-2}, x_{t-1})}
$$

Count-based estimates require smoothing to handle unseen contexts and may struggle to generalize. Neural language models learn distributed representations that can generalize across contexts, although they can still fail.

For mean negative log-likelihood $L$, perplexity is defined as:

$$
\boxed{\mathrm{PPL} = e^L}
$$

when natural logarithms are used, or:

$$
\boxed{\mathrm{PPL} = 2^L}
$$

when the loss uses base-2 logarithms.

**Interpretation:** Lower perplexity indicates better predictive probability on the **same tokenizer and evaluation corpus**. However, lower perplexity does not necessarily guarantee better factual accuracy or downstream task performance.


### The Pretraining Objective and Scaling Laws

For a causal language model, the standard pretraining objective is to minimize the **next-token negative log-likelihood** (equivalently, token-level cross-entropy):

$$
\boxed{\mathcal{L}_{\text{NLL}}(\theta)
= -\frac{1}{N}\sum_{t=1}^{N}\log P_{\theta}(x_t\mid x_{<t})}
$$

Here, $x_{<t}$ denotes the permitted preceding context and $N$ is the number of target token positions included in the loss. Training adjusts model parameters $\theta$ to assign higher probability to observed next tokens. This objective does **not** directly optimize every capability users care about, such as factuality, instruction following, or safety; these require appropriate data, post-training, and evaluation.

**Scaling laws** describe empirical relationships among model size, training-token count, compute, and loss. They are useful for planning, but they are not universal laws that predict application quality exactly. For example, the Chinchilla study showed that, under its compute-optimal training setup, both parameter count and token count mattered and many earlier models were undertrained. The right balance depends on the compute budget, data quality, architecture, and intended use.

- [Hoffmann et al., *Training Compute-Optimal Large Language Models* (Chinchilla)](https://arxiv.org/abs/2203.15556)

**Key takeaway:** Pretraining teaches a model to predict tokens from context. Low training loss is evidence that the objective is being optimized, not proof that a model is truthful, safe, or useful for every downstream task.



````python
mini_corpus = "the cat sat on the mat the cat ate fish the dog sat near the cat"
tokens_simple = mini_corpus.split()
context_counts = Counter(zip(tokens_simple[:-1], tokens_simple[1:]))
next_counts = defaultdict(Counter)
for a, b in zip(tokens_simple[:-1], tokens_simple[1:]):
    next_counts[a][b] += 1

def next_word_distribution(previous_word: str):
    counts = next_counts.get(previous_word, Counter())
    total = sum(counts.values())
    return {w: c/total for w,c in counts.items()} if total else {}

print("P(next | the) in tiny count corpus:", next_word_distribution("the"))
example_nll = -np.log([0.8, 0.5, 0.25]).mean()
print(f"Illustrative mean NLL={example_nll:.3f}; perplexity={np.exp(example_nll):.3f}")
````

### Training Alignment: Inputs vs. Targets

In autoregressive language modeling, the model learns to predict the **next token** from the tokens that came before it.

Consider the tokenized sentence:

$$
\boxed{[A,\ B,\ C,\ D]}
$$

The training sequence is shifted by one position to create inputs and targets:

$$
\boxed{
\begin{aligned}
\text{Inputs:}  &\quad [A,\ B,\ C] \\
\text{Targets:} &\quad [B,\ C,\ D]
\end{aligned}
}
$$

Each input position is trained to predict the token immediately following it:

| Position | Input token | Target token |
|---:|:---:|:---:|
| 0 | $A$ | $B$ |
| 1 | $B$ | $C$ |
| 2 | $C$ | $D$ |

**Key idea:** Position 0 predicts $B$, position 1 predicts $C$, and position 2 predicts $D$.

Shifting the sequence in the wrong direction can create information leakage or teach the model the wrong prediction task. The target alignment and causal attention mask must work together to ensure that predictions use only permitted context.

### Common Mistakes to Avoid

> **1. Mixing perplexities from different tokenizers**
>
> Perplexity depends on tokenization, so values from different tokenizers are not directly comparable without careful consideration.
>
> **2. Evaluating on training text**
>
> Evaluation on training examples does not provide an unbiased estimate of performance on unseen data.
>
> **3. Assuming token probabilities are independent**
>
> Autoregressive language models condition each next-token prediction on the preceding context; the sequence probability is factorized into conditional probabilities.
>
> **4. Forgetting the end-of-sequence token**
>
> The EOS token, when used, must be handled consistently with the vocabulary, training targets, and generation-stopping convention.



**Takeaway:** Correct next-token alignment is fundamental to causal language-model training. Always verify the input-target shift, prevent future-token leakage, and evaluate using a consistent tokenizer and held-out data.

## 6. Embeddings & Positional Information

<a id="chapter-6"></a>

### Why Do Transformers Need Positional Information?

**Token embeddings** represent the content of tokens as vectors. However, self-attention alone does not inherently know the order in which tokens appear.

Consider these two sentences:

> **Sentence A:** The dog chased the cat.

> **Sentence B:** The cat chased the dog.

They contain similar tokens, but their meanings differ because the token order changes. A Transformer therefore needs a mechanism to represent positional information alongside token content.

### How Content and Position Work Together?

A simplified processing pipeline is:

$$
\boxed{
\begin{gathered}
\text{Token IDs} \\
\downarrow \\
\text{Token Embeddings} \\
\downarrow \\
\text{Combine with Position Information} \\
\downarrow \\
\text{Self-Attention Layers} \\
\downarrow \\
\text{Contextual Representations}
\end{gathered}
}
$$

The exact method for incorporating position depends on the architecture.

### Three Common Approaches to Positional Information

| Method | How it works | Key characteristic |
|---|---|---|
| **Learned absolute embeddings** | Learns a vector for each position and combines it with the token representation | Flexible, but the trained position range may limit longer-context use |
| **Sinusoidal absolute encodings** | Uses fixed sine and cosine functions at multiple frequencies | Does not require learned positional parameters |
| **Relative methods / RoPE** | Encodes positional relationships through position-dependent transformations; RoPE rotates query and key feature pairs | Allows attention scores to reflect relative token offsets |

### 1. Learned Absolute Position Embeddings

Each token receives a content representation and a learned vector associated with its position.

A simplified representation is:

$$
\boxed{
h_i = e_i + p_i
}
$$

where:

- $e_i$ is the token embedding at position $i$.
- $p_i$ is the learned positional embedding for position $i$.
- $h_i$ is the resulting representation supplied to subsequent layers.

The positional vectors are learned during training. A checkpoint may support only a fixed trained range of positions unless an appropriate extension method is applied.

### 2. Sinusoidal Absolute Position Encodings

Sinusoidal encodings use sine and cosine functions at different frequencies to represent positions.

For model dimension $d_{\text{model}}$, the encoding is defined by:

$$
\boxed{
\operatorname{PE}(pos,2i)
=
\sin\left(
\frac{pos}{10000^{2i/d_{\text{model}}}}
\right)
}
$$

$$
\boxed{
\operatorname{PE}(pos,2i+1)
=
\cos\left(
\frac{pos}{10000^{2i/d_{\text{model}}}}
\right)
}
$$

Here:

- $pos$ is the token position.
- $i$ indexes the frequency pair.
- $d_{\text{model}}$ is the embedding dimension.
- Even dimensions use sine, while odd dimensions use cosine.

Different frequency scales allow the representation to encode positional patterns at different granularities.

### 3. Relative Position Methods and Rotary Position Embeddings (RoPE)

Relative position methods encode information about the relationship between token positions rather than relying only on an absolute position vector.

**Rotary Position Embeddings (RoPE)** achieve this by rotating pairs of query and key features according to their positions.

The position-dependent query and key representations can be written as:

$$
\boxed{
q_m' = R_m q_m,
\qquad
k_n' = R_n k_n
}
$$

where $R_m$ and $R_n$ are position-dependent rotation transformations for positions $m$ and $n$.

Their attention compatibility is then:

$$
\boxed{
\langle R_m q_m,\; R_n k_n \rangle
}
$$

Because the relative rotation depends on the difference between positions, the attention score can encode relative positional information.

````python
def sinusoidal_encoding(max_len: int, d_model: int):
    pos = np.arange(max_len)[:, None]
    i = np.arange(d_model)[None, :]
    angle_rates = 1 / np.power(10000, (2 * (i // 2)) / d_model)
    angles = pos * angle_rates
    pe = np.zeros((max_len, d_model), dtype=np.float32)
    pe[:, 0::2] = np.sin(angles[:, 0::2])
    pe[:, 1::2] = np.cos(angles[:, 1::2])
    return pe

pe = sinusoidal_encoding(80, 32)
plt.figure(figsize=(9,4)); plt.imshow(pe.T, aspect='auto', origin='lower', cmap='coolwarm')
plt.colorbar(label='Encoding value'); plt.xlabel('Position'); plt.ylabel('Embedding dimension')
plt.title('Sinusoidal positional encoding (computed from formula)'); plt.show()
print("Positional encoding shape:", pe.shape)
````

````python
# A minimal RoPE illustration for one 2D pair: rotate a vector by position-dependent angle.
def rotate_2d(vec: np.ndarray, theta: float) -> np.ndarray:
    R = np.array([[np.cos(theta), -np.sin(theta)], [np.sin(theta), np.cos(theta)]])
    return R @ vec

v = np.array([1.0, 0.0])
angles = np.linspace(0, 2*np.pi, 9)
rotated = np.stack([rotate_2d(v, a) for a in angles])
plt.figure(figsize=(4,4)); plt.quiver(np.zeros(len(rotated)), np.zeros(len(rotated)), rotated[:,0], rotated[:,1], angles='xy', scale_units='xy', scale=1)
plt.xlim(-1.2,1.2); plt.ylim(-1.2,1.2); plt.gca().set_aspect('equal'); plt.grid(alpha=.3)
plt.title('Toy rotary embedding: position changes rotation angle'); plt.xlabel('Feature 1'); plt.ylabel('Feature 2'); plt.show()
````

````python
# Apply rotary position embeddings to real query/key tensors shaped [B, H, T, D].
def apply_rope(x: torch.Tensor, base: float = 10000.0) -> torch.Tensor:
    """Educational RoPE implementation for even head dimensions; input shape [B,H,T,D]."""
    if x.ndim != 4:
        raise ValueError("Expected x with shape [B, H, T, D]")
    batch, heads, seq_len, dim = x.shape
    if dim % 2:
        raise ValueError("RoPE requires an even head dimension")
    positions = torch.arange(seq_len, device=x.device, dtype=x.dtype)[:, None]
    inverse_freq = 1.0 / (base ** (torch.arange(0, dim, 2, device=x.device, dtype=x.dtype) / dim))
    angles = positions * inverse_freq[None, :]
    cos = angles.cos()[None, None, :, :]
    sin = angles.sin()[None, None, :, :]
    even, odd = x[..., 0::2], x[..., 1::2]
    rotated = torch.empty_like(x)
    rotated[..., 0::2] = even * cos - odd * sin
    rotated[..., 1::2] = even * sin + odd * cos
    return rotated

torch.manual_seed(SEED)
q_demo = torch.randn(2, 4, 6, 8, device=DEVICE)
k_demo = torch.randn(2, 4, 6, 8, device=DEVICE)
q_rope, k_rope = apply_rope(q_demo), apply_rope(k_demo)
print("Q/K input shapes:", tuple(q_demo.shape), tuple(k_demo.shape))
print("Rotated Q/K shapes:", tuple(q_rope.shape), tuple(k_rope.shape))
print("Attention-score shapes before/after RoPE:", tuple((q_demo @ k_demo.transpose(-2, -1)).shape), tuple((q_rope @ k_rope.transpose(-2, -1)).shape))
print("RoPE preserves each adjacent pair's squared norm (max error):",
      float(((q_demo[..., 0::2]**2 + q_demo[..., 1::2]**2) -
             (q_rope[..., 0::2]**2 + q_rope[..., 1::2]**2)).abs().max().item()))
````

### Important Implementation Considerations

These approaches are **not interchangeable implementation details**. They affect how position is represented, how models behave outside their training context, numerical behavior, and checkpoint compatibility.

- **Learned absolute embeddings:** Depend on learned positional vectors and their trained range.
- **Sinusoidal encodings:** Use fixed mathematical functions rather than learned positional vectors.
- **RoPE:** Applies position-dependent rotations to query and key features, making relative offsets relevant to attention scores.

RoPE also has multiple scaling and context-extension variants. Supporting a longer context window does not automatically guarantee reliable performance over that range. The outcome depends on the positional method, training setup, inference configuration, and checkpoint.

**Key takeaway:** Token embeddings describe *what* a token represents, while positional information helps the model account for *where* it occurs in a sequence. The selected positional method influences how the Transformer interprets token relationships and handles longer contexts.

**Shape check:** token IDs `[B,T]` → token embeddings `[B,T,C]`; positional features must be broadcastable to `[B,T,C]` before addition for absolute encodings. RoPE is applied to compatible portions of query/key tensors, often shaped `[B,H,T,d_k]`.

## 7. Attention Mechanism

### Why Attention?

A recurrent model compresses prior tokens into a hidden state. In contrast, **self-attention** allows a token to compute a weighted mixture of representations from other tokens within its allowed context.

This provides **content-dependent information routing**. However, attention weights should not automatically be treated as faithful explanations of a model's reasoning.

Each token vector is projected into three representations:

- **Query ($Q$):** Represents what a token is looking for.
- **Key ($K$):** Represents what a token offers for matching.
- **Value ($V$):** Contains the information that can be aggregated.

The projections are computed as:

$$
\boxed{Q = XW_Q, \qquad K = XW_K, \qquad V = XW_V}
$$

### Scaled Dot-Product Attention

The attention operation is defined as:

$$
\boxed{\operatorname{Attention}(Q,K,V)
=
\operatorname{softmax}
\left(
\frac{QK^\top}{\sqrt{d_k}} + M
\right)V}
$$

where:

- $QK^\top$ measures the compatibility between queries and keys.
- $d_k$ is the key-vector dimension.
- $\sqrt{d_k}$ is the scaling factor that helps prevent dot products from becoming excessively large as the feature dimension increases.
- $M$ is an optional attention mask. Disallowed positions can be assigned $-\infty$ before applying softmax.
- $\operatorname{softmax}$ normalizes the scores across keys to produce attention weights.
- The resulting weights determine how the value vectors are combined.

### Tensor Shapes

Suppose the input tensor has shape:

$$
\boxed{X \in \mathbb{R}^{B \times T \times C}}
$$

where:

- $B$: Batch size.
- $T$: Sequence length.
- $C$: Model embedding dimension.

For a single attention head, the typical tensor shapes are:

| Tensor | Shape | Meaning |
|---|---|---|
| $Q, K, V$ | $(B,T,d_k)$ | Query, key, and value representations |
| $QK^\top$ | $(B,T,T)$ | Attention scores |
| Attention weights | $(B,T,T)$ | Normalized attention distributions |
| Head output | $(B,T,d_k)$ | Weighted combination of values |

### Multi-Head Attention

**Multi-head attention** runs several attention heads in parallel. Each head can learn different relationships between tokens.

The head outputs are concatenated and projected back to the model's embedding dimension:

$$
\boxed{\operatorname{MultiHead}(Q,K,V)
=
\operatorname{Concat}(H_1,\ldots,H_h)W_O}
$$

where $H_i$ is the output of the $i$-th attention head, $h$ is the number of heads, and $W_O$ is the output projection matrix.


````python
# A real, small numerical attention computation; values are generated reproducibly.
torch.manual_seed(SEED)
B, T, C, HEADS = 1, 5, 16, 4
D_HEAD = C // HEADS
X_demo = torch.randn(B, T, C)
Wq, Wk, Wv = [torch.randn(C, C) / np.sqrt(C) for _ in range(3)]
Q = (X_demo @ Wq).view(B, T, HEADS, D_HEAD).transpose(1, 2)  # [B,H,T,d_k]
K = (X_demo @ Wk).view(B, T, HEADS, D_HEAD).transpose(1, 2)
V = (X_demo @ Wv).view(B, T, HEADS, D_HEAD).transpose(1, 2)
scores = Q @ K.transpose(-2, -1) / np.sqrt(D_HEAD)            # [B,H,T,T]
weights = scores.softmax(dim=-1)
attn_out = weights @ V
print("X:", tuple(X_demo.shape), "Q/K/V:", tuple(Q.shape), "scores:", tuple(scores.shape), "output:", tuple(attn_out.shape))
print("Each row of attention weights sums to 1 (per head):", weights[0,0].sum(-1).detach().numpy().round(4))
````

````python
# Compare unrestricted attention and causal attention on the same scores.
causal_mask = torch.triu(torch.ones(T, T, dtype=torch.bool), diagonal=1)
causal_scores = scores.masked_fill(causal_mask, float('-inf'))
causal_weights = causal_scores.softmax(dim=-1)

fig, axes = plt.subplots(1, 2, figsize=(10,4))
axes[0].imshow(weights[0,0].detach().numpy(), vmin=0, vmax=1, cmap='viridis')
axes[0].set_title('Unmasked attention (one head)');
axes[1].imshow(causal_weights[0,0].detach().numpy(), vmin=0, vmax=1, cmap='viridis')
axes[1].set_title('Causal attention (one head)')
for ax in axes:
    ax.set_xlabel('Key / source position'); ax.set_ylabel('Query / current position')
    ax.set_xticks(range(T)); ax.set_yticks(range(T))
fig.colorbar(axes[1].images[0], ax=axes, label='Attention weight', shrink=.8)
fig.suptitle('Attention matrices computed from actual tensors'); fig.tight_layout(); plt.show()
print("Max attention weight above causal diagonal:", float(causal_weights[0,0].masked_select(causal_mask).max().item()) if causal_weights[0,0].masked_select(causal_mask).numel() else 0.0)
````

### Causal Masks & Padding Masks

- A **causal mask** blocks future keys from a query position. It is essential for next-token prediction so position `t` cannot see token `t+1` while learning to predict it.
- A **padding mask** prevents padded positions from being treated as real tokens in a batch.
- Mask polarity differs across APIs (some use `True` to block, others use `True` to keep); verify the specific API.

### Multi-head Attention

Each head has its own projection matrices and may learn different interaction patterns. Concatenating head outputs yields `[..., H*d_k]`, then an output projection returns `[..., C]`. More heads are not automatically better; the budget is shared across head dimension, model width, data and compute.

### Complexity Caveat

Dense self-attention forms a `T × T` interaction matrix, producing quadratic sequence-length compute/memory in the standard formulation. FlashAttention-style kernels can greatly reduce memory traffic and avoid materializing the full matrix in high-bandwidth memory, but they do not remove the fundamental dense all-pairs interaction count.

## 8. The Transformer Architecture

A canonical **Transformer block** contains self-attention, a position-wise feed-forward network, residual connections, and normalization.

Modern decoder-only models commonly use **pre-normalization** and may incorporate variants such as RMSNorm, gated MLPs (e.g., SwiGLU), rotary position embeddings, and optimized attention kernels. There is no single exact block that describes every Transformer implementation.

### A Simplified Pre-Norm Transformer Block

The two main sublayers can be represented as:

$$
\boxed{U = X + \operatorname{MHA}(\operatorname{Norm}(X))}
$$

$$
\boxed{Y = U + \operatorname{MLP}(\operatorname{Norm}(U))}
$$

where:

- $X$: Input representation to the Transformer block.
- $\operatorname{Norm}$: Normalization operation, such as LayerNorm or RMSNorm.
- $\operatorname{MHA}$: Multi-head attention.
- $\operatorname{MLP}$: Position-wise feed-forward network.
- $U$: Intermediate representation after the attention sublayer and residual connection.
- $Y$: Final output representation after the MLP sublayer and residual connection.

### Understanding the Components

- **Feed-forward network (MLP):** Often expands the representation from dimension $C$ to a wider hidden dimension, then projects it back to $C$.
- **Residual connections:** Provide identity paths for information and gradients, helping information flow through deeper networks.
- **Normalization:** Helps control activation scales and stabilize training.
- **Dropout:** Can regularize training by randomly dropping activations, although its use varies across large language models.

**Key takeaway:** Self-attention enables tokens to exchange information, the MLP transforms each token's representation, residual connections preserve information flow, and normalization helps stabilize the overall computation.


````python
# Executable pre-norm causal Transformer block, with shape and causality checks.
class DemoPreNormDecoderBlock(nn.Module):
    def __init__(self, d_model: int = 32, n_heads: int = 4, ff_multiplier: int = 4):
        super().__init__()
        if d_model % n_heads != 0:
            raise ValueError("d_model must be divisible by n_heads")
        self.norm1 = nn.LayerNorm(d_model)
        self.attn = nn.MultiheadAttention(d_model, n_heads, batch_first=True)
        self.norm2 = nn.LayerNorm(d_model)
        self.mlp = nn.Sequential(
            nn.Linear(d_model, ff_multiplier * d_model),
            nn.GELU(),
            nn.Linear(ff_multiplier * d_model, d_model),
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        # PyTorch Boolean attn_mask=True means the position is blocked.
        seq_len = x.shape[1]
        causal_mask = torch.triu(
            torch.ones(seq_len, seq_len, dtype=torch.bool, device=x.device), diagonal=1
        )
        normalized = self.norm1(x)
        attn_out, _ = self.attn(
            normalized, normalized, normalized,
            attn_mask=causal_mask, need_weights=False
        )
        x = x + attn_out
        return x + self.mlp(self.norm2(x))

block_demo = DemoPreNormDecoderBlock(d_model=32, n_heads=4).to(DEVICE).eval()
torch.manual_seed(SEED + 1)
x_block = torch.randn(2, 7, 32, device=DEVICE)
with torch.no_grad():
    y_block = block_demo(x_block)
    x_changed = x_block.clone()
    x_changed[:, -1, :] += 100.0  # Perturb only the final token.
    y_changed = block_demo(x_changed)
causal_difference = (y_block[:, :-1, :] - y_changed[:, :-1, :]).abs().max().item()
print("Input shape:", tuple(x_block.shape), "| block output shape:", tuple(y_block.shape))
print("Trainable parameters in this demonstration block:", sum(p.numel() for p in block_demo.parameters() if p.requires_grad))
print("Max change to earlier outputs after perturbing the final token:", f"{causal_difference:.8f}")
assert y_block.shape == x_block.shape
assert causal_difference < 1e-5, "Causal mask check failed: future token influenced earlier output."
print("Causal masking check: PASSED")
print("This small block is an executable demonstration; the later TinyGPT implementation remains the end-to-end decoder model.")
````

### Tensor-shape walkthrough

For input token IDs `idx [B,T]`:

1. Token embedding: `[B,T] → [B,T,C]`.
2. Attention projections: `[B,T,C] → Q,K,V [B,H,T,d_k]`, typically where `C=H*d_k`.
3. Attention scores: `Q @ Kᵀ → [B,H,T,T]`.
4. Weighted values: `[B,H,T,T] @ [B,H,T,d_k] → [B,H,T,d_k]`.
5. Concatenate heads: `[B,T,H*d_k] = [B,T,C]`.
6. MLP: `[B,T,C] → [B,T,C_ff] → [B,T,C]`.
7. Vocabulary projection: `[B,T,C] → logits [B,T,V]`.

Residual additions require compatible shapes. Cross-attention is different from self-attention because queries come from decoder states while keys and values come from encoder outputs.

## 9. Why Decoder-Only Transformers Power Many LLMs

Decoder-only models are a natural fit for text continuation because each position predicts the next token using a **causal mask**. During training, inputs and targets are shifted by one token, allowing the model to calculate losses for all token positions in parallel.

During ordinary autoregressive inference, however, tokens must generally be generated sequentially. Techniques such as batching and speculative decoding can improve throughput.

### Context Window & KV Cache

The **context window** defines how many tokens a particular model configuration can support as context during inference.

It is not an unlimited memory of previous interactions. Information may be truncated, forgotten, or poorly utilized even when it remains inside the context window.

A **key-value (KV) cache** stores attention keys and values from previously processed tokens. This avoids recomputing all previous key and value projections at every decoding step, making autoregressive generation more efficient.

### Comparing Transformer Architectures

| Architecture | Typical strengths | Example tasks | Important caveat |
|---|---|---|---|
| **Encoder-only** | Bidirectional contextual representations | Classification, embeddings, token labeling | Not designed as a native left-to-right text generator |
| **Decoder-only** | Text continuation and instruction-following after adaptation | Chat, code completion, text generation | Autoregressive token generation is sequential |
| **Encoder–decoder** | Conditional input-to-output mapping | Translation, summarization, structured transformation | Uses separate encoder and decoder components, typically connected through cross-attention |

**Key takeaway:** Architecture selection depends on the task, data, computational resources, and serving constraints—not on a universal ranking of architectures.

## 10. Dataset Preparation

### Understanding the Training Setup

We will train a **tiny character-level GPT-like decoder** on a short, original teaching corpus created inside the notebook.

The corpus is deliberately small and repetitive so training remains feasible on a local machine. It is **not a benchmark, a real-world dataset, or a source of meaningful general knowledge**.

### The Data Preparation Pipeline

$$
\boxed{
\begin{gathered}
\text{Collect Permitted Text} \\
\downarrow \\
\text{Normalize Text (as Appropriate)} \\
\downarrow \\
\text{Tokenize into Characters} \\
\downarrow \\
\text{Split Chronologically} \\
\downarrow \\
\text{Create Input--Target Batches} \\
\downarrow \\
\text{Train the Next-Token Predictor}
\end{gathered}
}
$$

Each stage prepares the text for autoregressive language modeling.

| Stage | Purpose |
|---|---|
| **Text collection** | Provides permitted text for the teaching corpus |
| **Normalization** | Standardizes text only where appropriate |
| **Character tokenization** | Maps individual characters to integer token IDs |
| **Chronological splitting** | Separates training, validation, and test portions while respecting sequence order |
| **Batch creation** | Groups fixed-length input sequences for model training |
| **Target shifting** | Uses the next character as the prediction target for each input position |

### How Next-Character Prediction Works

For an input sequence of tokens, the model learns to predict the token that follows each position.

$$
\boxed{
\begin{aligned}
\text{Input:}\quad & [x_1,x_2,\ldots,x_{T-1}] \\
\text{Target:}\quad & [x_2,x_3,\ldots,x_T]
\end{aligned}
}
$$

The target is shifted one position to the left relative to the original sequence. Each input position is supervised using its corresponding next token.

**Data-leakage precaution:** In rigorous projects, the vocabulary must be derived from the training split. The teaching demo uses a compact corpus and makes its simplifications explicit. Validation and test target text must not leak into training windows.

### Why Use Character-Level Tokenization?

**Advantages**

- Characters are easy to inspect and map to integer IDs.
- The vocabulary is small and conceptually straightforward.
- Tokenization can be implemented without downloading a pretrained tokenizer.

**Limitations**

- Character sequences are usually much longer than subword-token sequences.
- The model has no useful pretrained prior.
- Learning meaningful language patterns from a tiny corpus is inherently limited.

### Preparing Real-World Training Data

The small corpus in this notebook intentionally avoids the complexity of a production training dataset. For a real project, data preparation is a major source of both model quality and risk.

| Concern | Practical handling | Why it matters |
|---|---|---|
| **Provenance and permission** | Record source, license or permission basis, collection date, and allowed uses | Data availability does not automatically imply permission for every use |
| **Quality filtering** | Remove corrupt, irrelevant, spam-like, or low-quality records using documented rules | Low-quality data can consume compute and teach unwanted patterns |
| **Deduplication** | Detect exact and near-duplicate documents before splitting | Repetition can distort data balance and increase train/evaluation leakage |
| **Privacy and sensitive data** | Apply appropriate detection, minimization, and access controls | Training can preserve or reproduce sensitive material |
| **Evaluation isolation** | Keep benchmark and test examples out of training and tuning data | Leakage can make reported performance misleading |
| **Document boundaries** | Preserve end-of-document markers and avoid unintended cross-document targets | Concatenation without boundaries can create artificial context relationships |
| **Reproducibility** | Version datasets, filtering rules, tokenizer, and split manifests | Results need to be traceable to the exact data pipeline |

For large corpora, long documents are commonly tokenized and then packed into fixed-length training sequences to improve utilization. Packing must respect the intended document-boundary and attention semantics; otherwise, one document may accidentally condition on another. Splits should be created at an appropriate unit (for example, document, user, source, or time period), not merely by randomly splitting adjacent windows from the same document.

**Important distinction:** A tokenizer vocabulary or tokenizer configuration should be fixed consistently for the model and the entire evaluation pipeline. The tiny character-level demonstration makes vocabulary construction easy, but production models normally use a tokenizer trained and versioned separately.


````python
CORPUS = """
Language models predict tokens from context. A token can be a character, a word piece, or another symbol.
Attention combines information from earlier positions. A causal mask blocks future positions during next token training.
Small examples are useful for learning but do not behave like large assistants. Data quality and compute both matter.
A good experiment measures loss on held out text and inspects generated sequences with care.
Transformers use token embeddings, position information, attention, feed forward layers and residual connections.
A language model learns statistical structure; it does not guarantee truth, fairness, or reliable reasoning.
""".strip().lower()

# Use a stable character vocabulary for this teaching model.
chars = sorted(set(CORPUS))
char_to_id = {ch: i for i, ch in enumerate(chars)}
id_to_char = {i: ch for ch, i in char_to_id.items()}
encoded_corpus = torch.tensor([char_to_id[ch] for ch in CORPUS], dtype=torch.long)
split_idx = int(0.9 * len(encoded_corpus))
train_data = encoded_corpus[:split_idx]
val_data = encoded_corpus[split_idx:]
VOCAB_SIZE = len(chars)
print(f"Corpus chars: {len(CORPUS)} | vocab: {VOCAB_SIZE} | train: {len(train_data)} | validation: {len(val_data)}")
print("Vocabulary excerpt:", repr(''.join(chars[:40])))
````

````python
BATCH_SIZE = 24
BLOCK_SIZE = 48

def get_batch(split: str, batch_size: int = BATCH_SIZE, block_size: int = BLOCK_SIZE):
    """Return x,y token windows with next-character alignment; x,y shapes [B,T]."""
    data = train_data if split == 'train' else val_data
    if len(data) <= block_size + 1:
        raise ValueError(f"Need more {split} tokens than block_size+1; got {len(data)} tokens")
    starts = torch.randint(0, len(data) - block_size - 1, (batch_size,))
    x = torch.stack([data[i:i+block_size] for i in starts])
    y = torch.stack([data[i+1:i+block_size+1] for i in starts])
    return x.to(DEVICE), y.to(DEVICE)

xb, yb = get_batch('train')
print("Input batch shape:", tuple(xb.shape), "target batch shape:", tuple(yb.shape))
print("First input:", ''.join(id_to_char[int(i)] for i in xb[0].cpu()))
print("Target is input shifted one character:", ''.join(id_to_char[int(i)] for i in yb[0].cpu()))
assert xb.shape == yb.shape == (BATCH_SIZE, BLOCK_SIZE)
````

## 11. Implement a GPT-Style Model in PyTorch

### From Tokens to Next-Token Predictions

The next code cell builds the architecture using primitive PyTorch layers. It **does not load a pretrained GPT model**.

The implementation combines token and position embeddings with masked self-attention, a feed-forward network, residual connections, LayerNorm, and a vocabulary projection.

### Architecture at a Glance

$$
\boxed{
\begin{gathered}
\text{Token IDs} \\
\downarrow \\
\text{Token Embeddings} + \text{Position Embeddings} \\
\downarrow \\
\text{Masked Multi-Head Self-Attention} \\
\downarrow \\
\text{Residual Connection and LayerNorm} \\
\downarrow \\
\text{Feed-Forward Network} \\
\downarrow \\
\text{Residual Connection and LayerNorm} \\
\downarrow \\
\text{Vocabulary Logits} \\
\downarrow \\
\text{Next-Token Prediction}
\end{gathered}
}
$$

### Architectural Choices

For clarity, this model uses:

- **Learned absolute position embeddings** to represent token positions.
- **LayerNorm** to normalize hidden representations.
- **GELU** as the feed-forward activation.
- **Ordinary masked self-attention** to exchange information across permitted positions.

This is a **GPT-like teaching implementation**, not an exact reproduction of any particular production checkpoint.


````python
class CausalSelfAttention(nn.Module):
    def __init__(self, n_embd: int, n_head: int, block_size: int, dropout: float = 0.0):
        super().__init__()
        if n_embd % n_head != 0:
            raise ValueError("n_embd must be divisible by n_head")
        self.n_head = n_head
        self.head_dim = n_embd // n_head
        self.qkv = nn.Linear(n_embd, 3*n_embd, bias=False)
        self.proj = nn.Linear(n_embd, n_embd)
        self.attn_dropout = nn.Dropout(dropout)
        self.resid_dropout = nn.Dropout(dropout)
        mask = torch.tril(torch.ones(block_size, block_size, dtype=torch.bool))
        self.register_buffer('causal_mask', mask, persistent=False)

    def forward(self, x: torch.Tensor, return_weights: bool = False):
        B, T, C = x.shape
        q, k, v = self.qkv(x).chunk(3, dim=-1)
        q = q.view(B, T, self.n_head, self.head_dim).transpose(1, 2)
        k = k.view(B, T, self.n_head, self.head_dim).transpose(1, 2)
        v = v.view(B, T, self.n_head, self.head_dim).transpose(1, 2)
        scores = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)  # [B,H,T,T]
        mask = self.causal_mask[:T, :T]
        scores = scores.masked_fill(~mask, float('-inf'))
        weights = scores.softmax(dim=-1)
        attended = self.attn_dropout(weights) @ v
        attended = attended.transpose(1, 2).contiguous().view(B, T, C)
        out = self.resid_dropout(self.proj(attended))
        return (out, weights) if return_weights else out


class FeedForward(nn.Module):
    def __init__(self, n_embd: int, dropout: float = 0.0):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(n_embd, 4*n_embd),
            nn.GELU(),
            nn.Linear(4*n_embd, n_embd),
            nn.Dropout(dropout),
        )
    def forward(self, x):
        return self.net(x)


class TransformerBlock(nn.Module):
    def __init__(self, n_embd: int, n_head: int, block_size: int, dropout: float = 0.0):
        super().__init__()
        self.ln1 = nn.LayerNorm(n_embd)
        self.attn = CausalSelfAttention(n_embd, n_head, block_size, dropout)
        self.ln2 = nn.LayerNorm(n_embd)
        self.ff = FeedForward(n_embd, dropout)
    def forward(self, x, return_weights: bool = False):
        if return_weights:
            attn_out, weights = self.attn(self.ln1(x), return_weights=True)
        else:
            attn_out = self.attn(self.ln1(x))
        x = x + attn_out
        x = x + self.ff(self.ln2(x))
        return (x, weights) if return_weights else x


class TinyGPT(nn.Module):
    def __init__(self, vocab_size: int, block_size: int, n_embd: int = 64, n_head: int = 4, n_layer: int = 2, dropout: float = 0.05):
        super().__init__()
        self.block_size = block_size
        self.token_embedding = nn.Embedding(vocab_size, n_embd)
        self.position_embedding = nn.Embedding(block_size, n_embd)
        self.drop = nn.Dropout(dropout)
        self.blocks = nn.ModuleList([TransformerBlock(n_embd, n_head, block_size, dropout) for _ in range(n_layer)])
        self.ln_f = nn.LayerNorm(n_embd)
        self.lm_head = nn.Linear(n_embd, vocab_size, bias=False)
        # Weight tying is common in decoder LMs; share input embedding and output weights.
        self.lm_head.weight = self.token_embedding.weight
        self.apply(self._init_weights)

    @staticmethod
    def _init_weights(module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None: nn.init.zeros_(module.bias)
        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

    def forward(self, idx: torch.Tensor, targets: torch.Tensor | None = None, return_weights: bool = False):
        B, T = idx.shape
        if T > self.block_size:
            raise ValueError(f"Input length {T} exceeds block_size {self.block_size}")
        positions = torch.arange(T, device=idx.device)
        x = self.drop(self.token_embedding(idx) + self.position_embedding(positions)[None, :, :])
        last_weights = None
        for block in self.blocks:
            if return_weights:
                x, last_weights = block(x, return_weights=True)
            else:
                x = block(x)
        logits = self.lm_head(self.ln_f(x))  # [B,T,V]
        loss = None
        if targets is not None:
            loss = F.cross_entropy(logits.reshape(B*T, -1), targets.reshape(B*T))
        if return_weights:
            return logits, loss, last_weights
        return logits, loss

    @torch.no_grad()
    def generate(self, idx: torch.Tensor, max_new_tokens: int = 100, temperature: float = 1.0, top_k: int | None = None):
        self.eval()
        for _ in range(max_new_tokens):
            idx_cond = idx[:, -self.block_size:]
            logits, _ = self(idx_cond)
            logits = logits[:, -1, :] / max(temperature, 1e-6)
            if top_k is not None:
                k = min(top_k, logits.size(-1))
                threshold = torch.topk(logits, k).values[:, [-1]]
                logits = logits.masked_fill(logits < threshold, float('-inf'))
            probs = F.softmax(logits, dim=-1)
            next_id = torch.multinomial(probs, num_samples=1)
            idx = torch.cat([idx, next_id], dim=1)
        return idx

model = TinyGPT(VOCAB_SIZE, BLOCK_SIZE).to(DEVICE)
logits0, loss0 = model(xb, yb)
print("Parameter count:", sum(p.numel() for p in model.parameters()))
print("Logits shape:", tuple(logits0.shape), "loss:", float(loss0.item()))
assert logits0.shape == (BATCH_SIZE, BLOCK_SIZE, VOCAB_SIZE)
assert torch.isfinite(loss0)
````

````python
class KVCacheAttention(nn.Module):
    """Small multi-head attention layer exposing full-prefix and cached decode paths."""
    def __init__(self, d_model: int = 32, n_head: int = 4):
        super().__init__()
        if d_model % n_head:
            raise ValueError("d_model must be divisible by n_head")
        self.d_model, self.n_head = d_model, n_head
        self.head_dim = d_model // n_head
        self.qkv = nn.Linear(d_model, 3 * d_model, bias=False)
        self.out_proj = nn.Linear(d_model, d_model, bias=False)

    def _project(self, x):
        batch, seq_len, _ = x.shape
        q, k, v = self.qkv(x).chunk(3, dim=-1)
        def split_heads(t):
            return t.view(batch, seq_len, self.n_head, self.head_dim).transpose(1, 2)
        return split_heads(q), split_heads(k), split_heads(v)

    def _merge(self, x):
        batch, heads, seq_len, head_dim = x.shape
        return x.transpose(1, 2).contiguous().view(batch, seq_len, heads * head_dim)

    def forward_full(self, x):
        q, k, v = self._project(x)
        seq_len = x.shape[1]
        scores = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        causal = torch.ones(seq_len, seq_len, dtype=torch.bool, device=x.device).tril()
        scores = scores.masked_fill(~causal, float('-inf'))
        return self.out_proj(self._merge(scores.softmax(dim=-1) @ v))

    def prefill(self, x):
        """Process the prompt and return output plus cached keys/values."""
        q, k, v = self._project(x)
        seq_len = x.shape[1]
        scores = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        causal = torch.ones(seq_len, seq_len, dtype=torch.bool, device=x.device).tril()
        scores = scores.masked_fill(~causal, float('-inf'))
        output = self.out_proj(self._merge(scores.softmax(dim=-1) @ v))
        return output, (k, v)

    def decode_step(self, x_new, cache):
        """Process a new token; cached keys/values contain only past/current positions."""
        q, k_new, v_new = self._project(x_new)
        if cache is None:
            k_all, v_all = k_new, v_new
        else:
            k_all = torch.cat([cache[0], k_new], dim=2)
            v_all = torch.cat([cache[1], v_new], dim=2)
        scores = (q @ k_all.transpose(-2, -1)) / math.sqrt(self.head_dim)
        output = self.out_proj(self._merge(scores.softmax(dim=-1) @ v_all))
        return output, (k_all, v_all)


torch.manual_seed(SEED)
cache_attention_demo = KVCacheAttention(d_model=32, n_head=4).to(DEVICE).eval()
cache_input = torch.randn(2, 9, 32, device=DEVICE)
with torch.no_grad():
    full_outputs = cache_attention_demo.forward_full(cache_input)
    prefill_outputs, kv_cache = cache_attention_demo.prefill(cache_input[:, :3])
    incremental_outputs = [prefill_outputs]
    for position in range(3, cache_input.shape[1]):
        one_output, kv_cache = cache_attention_demo.decode_step(cache_input[:, position:position+1], kv_cache)
        incremental_outputs.append(one_output)
    cached_outputs = torch.cat(incremental_outputs, dim=1)
print("Full attention output:", tuple(full_outputs.shape))
print("Cached decode output:", tuple(cached_outputs.shape))
print("Final K/V cache shapes:", tuple(kv_cache[0].shape), tuple(kv_cache[1].shape))
print("Maximum full-vs-cached output difference:", float((full_outputs - cached_outputs).abs().max().item()))
assert torch.allclose(full_outputs, cached_outputs, atol=1e-5, rtol=1e-5), "Cached attention should match full causal attention."
````

### Reading the Implementation

| Component | What it does |
|---|---|
| **`qkv` projection** | Performs three linear projections in one fused layer and then splits the result into queries, keys, and values |
| **Attention heads** | Reshapes tensors into `[B, H, T, d_k]` so attention is calculated independently per head |
| **Causal mask** | Uses `masked_fill(~mask, -inf)` to prevent future positions from contributing to the attention softmax |
| **`FeedForward`** | Transforms each sequence position independently, while attention moves information across positions |
| **Residual connections** | Preserve the hidden-state shape `[B, T, C]` across the sublayers |
| **`lm_head`** | Maps hidden states to vocabulary logits with shape `[B, T, V]` |
| **Next-token targets** | Train each position against the next character ID |

Here, the tensor dimensions are:

$$
\boxed{
X \in \mathbb{R}^{B \times T \times C}
}
$$

where:

- $B$ is the batch size.
- $T$ is the sequence length.
- $C$ is the model embedding dimension.
- $H$ is the number of attention heads.
- $d_k$ is the dimension of each attention head.
- $V$ is the vocabulary size.

For the attention operation, the typical shapes are:

$$
\boxed{
\begin{aligned}
Q,K,V &\in \mathbb{R}^{B \times H \times T \times d_k} \\
\text{Attention Scores} &\in \mathbb{R}^{B \times H \times T \times T} \\
\text{Head Output} &\in \mathbb{R}^{B \times H \times T \times d_k}
\end{aligned}
}
$$

The vocabulary projection produces:

$$
\boxed{
\text{Logits} \in \mathbb{R}^{B \times T \times V}
}
$$

Each position produces a score for every character in the vocabulary. During training, these logits are compared with the appropriately shifted target character IDs.

## 12. Train & Inspect the Model

### Understanding the Training Process

The training loop uses a short, fixed training budget to demonstrate the mechanics of training a GPT-style model. The primary goal is not to achieve high-quality text generation. Instead, the experiment verifies that the training process works correctly and that the model can be inspected after training.

\text{Run Training Loop} \\
\downarrow \\
\text{Verify Finite Training Loss} \\

\text{Evaluate on Held-Out Data} \\

\text{Save Model Checkpoint} \\

\text{Inspect Training Behavior}

The notebook records actual results from the run rather than claiming predetermined loss values. Loss values can vary with the execution environment, random initialization, and training configuration.

````python
# Reset weights by creating a fresh model, then run a reproducible lightweight experiment.
torch.manual_seed(SEED)
model = TinyGPT(VOCAB_SIZE, BLOCK_SIZE, n_embd=64, n_head=4, n_layer=2, dropout=0.05).to(DEVICE)
optimizer = torch.optim.AdamW(model.parameters(), lr=2e-3, weight_decay=0.01)
TRAIN_STEPS = 180  # increase for a longer experiment; CPU runtime grows with model and sequence size
EVAL_EVERY = 30
train_losses, val_losses, eval_steps = [], [], []

def estimate_loss(eval_batches: int = 4):
    model.eval()
    result = {}
    with torch.no_grad():
        for split in ('train', 'val'):
            losses = []
            for _ in range(eval_batches):
                try:
                    ex, ey = get_batch(split)
                except ValueError:
                    # Only applies if a user edits the corpus/block size incompatibly.
                    return {"error": "Not enough tokens for selected block size."}
                _, loss_val = model(ex, ey)
                losses.append(loss_val.item())
            result[split] = float(np.mean(losses))
    model.train()
    return result

start_train = time.perf_counter()
for step in range(1, TRAIN_STEPS + 1):
    model.train()
    x_batch, y_batch = get_batch('train')
    _, loss_train = model(x_batch, y_batch)
    optimizer.zero_grad(set_to_none=True)
    loss_train.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
    optimizer.step()
    train_losses.append(loss_train.item())
    if step % EVAL_EVERY == 0 or step == 1:
        measured = estimate_loss()
        eval_steps.append(step)
        val_losses.append(measured.get('val', float('nan')))
        print(f"step {step:3d} | minibatch train loss {loss_train.item():.3f} | sampled validation loss {val_losses[-1]:.3f}")
train_seconds = time.perf_counter() - start_train
print(f"Training wall time on {DEVICE}: {train_seconds:.2f} seconds (measured in this run; varies by hardware).")

checkpoint_path = 'tiny_gpt_checkpoint.pt'
torch.save({'model_state_dict': model.state_dict(), 'vocab': chars, 'block_size': BLOCK_SIZE}, checkpoint_path)
print("Saved checkpoint:", checkpoint_path)
````

| Symptoms | What to check |
|---|---|
| **Loss becomes `NaN`** | Lower the learning rate, check the data and targets, inspect gradients, and verify that every row of the causal mask allows at least one key |
| **Model does not appear to learn** | Confirm that input and target sequences are shifted correctly and that optimizer steps are actually being executed |
| **Training loss decreases, but validation loss worsens** | Check for overfitting, limited dataset diversity, and differences between training and validation distributions |
| **Generated text is repetitive or incoherent** | Inspect corpus size, training duration, tokenization, and sampling configuration; poor output quality is expected from a tiny teaching model |

**Note:** The corpus is small and repetitive, so its validation loss and generated text should be interpreted as implementation diagnostics—not as evidence of general language understanding or competitive model performance.

### Key Takeaway

A successful training run is more than a decreasing loss curve. Verify that the objective is computed correctly, validation data remains held out, checkpoints are saved, and generation works. Then inspect the observed results to understand what the model has learned and where it fails.

### From a Laptop Experiment to Larger Training Runs

The training loop here is intentionally compact. Larger runs introduce additional controls for optimization stability, memory, throughput, and reproducibility.

| Technique | What it does | Trade-off or caution |
|---|---|---|
| **Learning-rate schedule and warm-up** | Changes the learning rate over training; warm-up can stabilize early updates | A schedule that is poorly matched to the token budget can slow or destabilize learning |
| **Gradient clipping** | Limits the gradient norm before an optimizer update | It can prevent extreme updates, but cannot repair incorrect data or loss logic |
| **Mixed precision (FP16/BF16)** | Uses lower-precision arithmetic for speed and memory savings | Numerical range and hardware support differ; monitor loss and stability |
| **Gradient accumulation** | Accumulates gradients across smaller micro-batches before updating | Increases effective batch size without reducing all per-step activation memory, but adds compute and changes update frequency |
| **Activation checkpointing** | Recomputes some activations during backward propagation instead of storing all of them | Saves memory at the cost of extra computation |
| **Distributed data parallelism** | Replicates a model across devices and synchronizes gradients over different data batches | Communication and optimizer memory remain important bottlenecks |
| **Tensor / pipeline parallelism** | Splits model computation or layers across devices | Adds communication and operational complexity |

A fair comparison should report the number of training tokens, context length, batch configuration, precision, optimizer and schedule, compute hardware, wall-clock time, and held-out metrics. Parameter count alone is not a complete measure of training effort or quality.

**Scaling-law caution:** Better loss from a larger model or longer run should be measured on held-out data. More parameters do not automatically compensate for poor data, too few training tokens, data leakage, or a mismatch between the pretraining objective and the target application. See the [Chinchilla scaling study](https://arxiv.org/abs/2203.15556) for one influential compute-optimal analysis; treat its findings as empirical guidance, not a universal recipe.

## 13. Text Generation & Sampling

### From Logits to Generated Text

During generation, the model produces **logits**—unnormalized scores over the vocabulary. A decoding strategy transforms these scores into a next-token choice. The generation process can be summarized as:

$$
\boxed{
\begin{gathered}
\text{Context Tokens} \\
\downarrow \\
\text{Model Produces Logits} \\
\downarrow \\
\text{Apply Sampling Strategy} \\
\downarrow \\
\text{Select the Next Token} \\
\downarrow \\
\text{Append Token to Context} \\
\downarrow \\
\text{Repeat Until Stopping}
\end{gathered}
}
$$

### Common Decoding Strategies

| Strategy | How it works | Main characteristic |
|---|---|---|
| **Greedy decoding** | Selects the most likely token at each step | Deterministic, but can produce repetitive sequences |
| **Temperature sampling** | Scales logits before computing probabilities | Lower temperatures sharpen the distribution; higher temperatures flatten it |
| **Top-k sampling** | Retains only the $k$ highest-logit candidates | Restricts choices to a fixed number of candidates |
| **Top-p (nucleus) sampling** | Retains the smallest high-probability set whose cumulative probability reaches $p$ | Adapts the number of candidates to the probability distribution |

### 1. Temperature

If $z_i$ is the logit for vocabulary token $i$, temperature-adjusted probabilities are computed as:

$$
\boxed{
P(x=i)
=
\frac{\exp(z_i/\tau)}
{\displaystyle\sum_j \exp(z_j/\tau)}
}
$$

where $\tau$ is the temperature parameter.

- **Lower temperature:** Produces a sharper probability distribution, concentrating probability on higher-scoring tokens.
- **Higher temperature:** Produces a flatter distribution, making lower-scoring tokens more likely to be sampled.
- **Important:** Temperature changes the sampling distribution; it does not make the model more knowledgeable.

### 2. Top-k Sampling

Top-k sampling retains only the $k$ candidates with the highest logits before sampling.

$$
\boxed{
\mathcal{V}_k
=
\operatorname{TopK}(z,k)
}
$$

The remaining candidates are excluded from sampling, and the probabilities of the retained candidates are renormalized.

### 3. Top-p (Nucleus) Sampling

Top-p sampling selects the smallest set of high-probability candidates whose cumulative probability reaches the threshold $p$. After sorting token probabilities in descending order, the retained set is approximately characterized by:

$$
\boxed{
\sum_{i \in \mathcal{V}_p} P(x=i) \geq p
}
$$

Implementation details, including whether the boundary-crossing token is retained, can vary. Unlike top-k, top-p adapts the number of candidates to the shape of the probability distribution.

````python
def top_p_filter(logits: torch.Tensor, p: float = 0.9):
    """Filter a 1D logit tensor by nucleus probability mass; leaves at least one token."""
    if logits.ndim != 1:
        raise ValueError("Expected one-dimensional logits")
    sorted_logits, sorted_indices = torch.sort(logits, descending=True)
    sorted_probs = F.softmax(sorted_logits, dim=-1)
    cumulative = torch.cumsum(sorted_probs, dim=-1)
    remove = cumulative > p
    remove[1:] = remove[:-1].clone()
    remove[0] = False
    filtered = logits.clone()
    filtered[sorted_indices[remove]] = float('-inf')
    return filtered

# Compare distribution transforms on actual logits from model output.
with torch.no_grad():
    base_logits = sample_logits[0, -1].detach().cpu()
temps = [0.5, 1.0, 1.7]
fig, axes = plt.subplots(1, 3, figsize=(12,3.6))
for ax, temp in zip(axes, temps):
    p = F.softmax(base_logits / temp, dim=-1).numpy()
    ax.bar(np.arange(len(p)), p)
    ax.set_title(f'Temperature τ={temp}'); ax.set_xlabel('Vocabulary ID'); ax.set_ylabel('Probability')
fig.suptitle('Temperature reshapes the same model logits'); fig.tight_layout(); plt.show()

filtered_top_p = top_p_filter(base_logits, p=0.85)
filtered_probs = F.softmax(filtered_top_p, dim=-1)
print("Base distribution support:", int((F.softmax(base_logits, dim=-1) > 0).sum()))
print("After top-p filtering support:", int((filtered_probs > 0).sum()))
````

````python
@torch.no_grad()
def greedy_generate(model, context: torch.Tensor, max_new_tokens: int = 120):
    model.eval(); idx = context
    for _ in range(max_new_tokens):
        logits, _ = model(idx[:, -model.block_size:])
        next_id = logits[:, -1, :].argmax(dim=-1, keepdim=True)
        idx = torch.cat((idx, next_id), dim=1)
    return idx

start_text = "language"
start_ids = torch.tensor([[char_to_id[ch] for ch in start_text if ch in char_to_id]], device=DEVICE)
greedy_ids = greedy_generate(model, start_ids, max_new_tokens=120)[0].cpu().tolist()
sample_ids = model.generate(start_ids, max_new_tokens=120, temperature=0.9, top_k=8)[0].cpu().tolist()
print("GREEDY OUTPUT:\n", ''.join(id_to_char[i] for i in greedy_ids))
print("\nSAMPLED OUTPUT (top-k=8):\n", ''.join(id_to_char[i] for i in sample_ids))
````

### Assessing the Result Honestly

Character-level text generated from this small corpus is expected to resemble fragments of familiar spelling patterns rather than coherent assistant responses. This is a **successful demonstration of the mechanism**, not a successful general-purpose LLM.

### Why Can the Generated Text Be Repetitive?

| Possible cause | Explanation |
|---|---|
| **Greedy decoding** | Repeatedly selecting the highest-probability token can lead to repetitive sequences |
| **Small dataset** | Limited training data restricts the patterns the model can learn |
| **Limited capacity** | A small model may struggle to capture complex linguistic structure |
| **Under-training** | Insufficient training may prevent the model from learning useful patterns |

These factors can interact. Changing the decoding strategy may alter the output, but it cannot compensate for missing training data or insufficient model capacity.

### Key Takeaway

Text generation combines a learned next-token probability distribution with a decoding strategy. Greedy, temperature, top-k, and top-p sampling control how the next token is selected, but the quality of the result still depends on the model, training data, and learned patterns.

A **pretrained language model** is a model whose parameters have already been learned from a training process. Instead of starting with randomly initialized weights, you load an existing checkpoint and use it for inference, fine-tuning, or further experimentation.

$$
\boxed{\text{Input Text}
\rightarrow
\text{Tokenizer}
\rightarrow
\text{Token IDs}
\rightarrow
\text{Model}
\rightarrow
\text{Output Token IDs}
\rightarrow
\text{Decoded Text}}
$$

**Key idea:** The tokenizer and model must be compatible. Using an incorrect tokenizer can produce invalid or unexpected results, even when the model weights load successfully.

### Understanding Model Size, Memory, and Model Documentation

#### 1. Estimating Raw Model Weight Memory

A model with $N$ parameters, where each parameter is stored using $b$ bits, requires approximately the following memory for its raw weights:

$$
\boxed{\text{Weight Memory (bytes)} \approx \frac{N \times b}{8}}
$$

For example, consider a model with $7$ billion parameters stored in 16-bit precision:

$$
\boxed{\begin{aligned}
N &= 7 \times 10^9 \\
b &= 16 \\
\text{Memory} &\approx \frac{7 \times 10^9 \times 16}{8} \\
&= 14 \times 10^9 \text{ bytes}
\end{aligned}}
$$

This is approximately **14 GB in decimal units**, or **13.04 GiB** in binary units, for raw weights alone.

**Important:** This estimate excludes metadata, activations, temporary buffers, the KV cache, optimizer states, memory fragmentation, and framework overhead.

#### 2. Why Training Requires More Memory?

Training is generally more memory-intensive than inference because it may require storing gradients, optimizer states, and intermediate activations in addition to model weights.

| Memory component | Purpose |
|---|---|
| **Model weights** | Store the learned parameters |
| **Gradients** | Record how the loss changes with respect to parameters |
| **Optimizer states** | Store additional information used by optimizers, such as AdamW's moving averages |
| **Activations** | Retain intermediate values needed for backpropagation |
| **KV cache** | Store attention keys and values during autoregressive inference |
| **Temporary buffers and overhead** | Support computation, memory management, and framework operations |

The exact memory requirements depend on the model architecture, numerical precision, optimizer, batch size, sequence length, and training or inference configuration.

#### 3. Understanding Model Specifications

| Quantity | What it tells you | What it does not tell you |
|---|---|---|
| **Parameter count** | Size of the collection of learned parameters | Quality, truthfulness, or inference speed by itself |
| **Context window** | Maximum configured input or working-context length | Perfect recall or unlimited conversational memory |
| **Model card** | Documented intended use, evaluations, and limitations | Independent proof that every claim is complete or accurate |
| **License** | Conditions governing use, modification, and distribution | Whether the model is safe or suitable for a particular task |

#### 4. Official Hugging Face References

For current APIs and implementation details, consult the official documentation:

- [Causal language modeling guide](https://huggingface.co/docs/transformers/en/tasks/language_modeling) — explains causal language-model training and fine-tuning.
- [Transformers pipeline documentation](https://huggingface.co/docs/transformers/en/main_classes/pipelines) — describes high-level interfaces for tasks such as text generation and classification.

**Key takeaway:** Parameter count provides a useful starting point for estimating model size, but it does not determine total memory requirements or model quality. Always consider the full execution setup and consult the model card, license, and official documentation before deploying a model.

## 15. Prompt Engineering

### What Is Prompt Engineering?

**Prompt engineering** is the process of designing the input provided to a language model to guide its response toward a desired task or outcome. Reliable prompts typically specify the task, provide relevant context, define constraints and output formats, and include examples when the task benefits from them. Prompting can improve consistency, but it does **not guarantee correctness** and should never replace systematic evaluation.

### A Practical Prompt Structure

A useful prompt can be organized into six components:

$$
\boxed{
\begin{gathered}
\text{Role / Task} \\
\downarrow \\
\text{Context} \\
\downarrow \\
\text{Requirements} \\
\downarrow \\
\text{Examples} \\
\downarrow \\
\text{Output Schema} \\
\downarrow \\
\text{Uncertainty Policy}
\end{gathered}
}
$$

| Component | Purpose | Example |
|---|---|---|
| **Role / Task** | Establishes what the model should do | "Act as a Python tutor and explain this function." |
| **Context** | Supplies relevant background information | "The code processes a CSV dataset." |
| **Requirements** | Defines constraints and expectations | "Explain each step and identify edge cases." |
| **Examples** | Demonstrates the desired response pattern | Provide an input and its expected output |
| **Output schema** | Specifies the required response structure | "Return a JSON object with `summary` and `issues`." |
| **Uncertainty policy** | Explains how to handle missing information | "State when the evidence is insufficient; do not invent facts." |

*Not every prompt needs all six components. Use the components that meaningfully improve the task.

### Zero-Shot vs. Few-Shot Prompting

| Technique | Description | When to use it |
|---|---|---|
| **Zero-shot prompting** | Gives instructions without examples | Straightforward tasks with clear instructions |
| **Few-shot prompting** | Includes examples of the desired input-output behavior | Tasks that benefit from demonstrated patterns or formatting |

### Key Takeaway

Good prompt engineering makes the task, context, constraints, and expected output explicit. Use examples when helpful, define how uncertainty should be handled, and combine prompting with evaluation and validation rather than treating a well-written prompt as a guarantee of correctness.

### Prompt injection & Boundaries

Prompt injection occurs when untrusted input attempts to manipulate a language model into ignoring its intended instructions or performing unintended actions.

Instructions and quoted content are not inherently trustworthy simply because they appear in a prompt. In Retrieval-Augmented Generation (RAG) or agent systems, a retrieved document may contain adversarial instructions, such as telling the model to ignore previous instructions or reveal sensitive information.

### How Prompt Injection Can Enter a System?

A typical attack path looks like this:

$$
\boxed{
\begin{gathered}
\text{User Query or Retrieved Document} \
\rightarrow \
\text{Adversarial Instructions} \
\rightarrow \
\text{Model May Follow Untrusted Content} \
\rightarrow \
\text{Potentially Unsafe Output or Action}
\end{gathered}
}
$$

This illustrates a potential risk, not an inevitable outcome. The impact depends on the model, application design, available tools, and security controls.

### A simple local structured-output validator


````python
import json

def validate_summary_payload(text: str):
    """Parse and validate a minimal expected JSON structure; do not trust output blindly."""
    try:
        obj = json.loads(text)
    except json.JSONDecodeError as exc:
        return False, f"Invalid JSON: {exc.msg}"
    required = {"behavior", "failure_handling", "observability"}
    if set(obj) != required:
        return False, f"Expected exactly keys {sorted(required)}"
    if not all(v is None or isinstance(v, str) for v in obj.values()):
        return False, "Every value must be a string or null"
    return True, obj

candidate = '{"behavior":"Retries failed jobs three times","failure_handling":"Writes a dead-letter record after final failure","observability":"Exposes a retry metric"}'
print(validate_summary_payload(candidate))
print(validate_summary_payload('{"behavior": true}'))
````

## 16. Fine-Tuning & Instruction Tuning

### What Is Fine-Tuning?

**Fine-tuning** continues training a model from its pretrained weights using a target dataset. It adapts the model's existing capabilities to a particular task, domain, response style, or output format.

**Supervised fine-tuning (SFT)** trains the model to reproduce desired target responses for supplied instructions or inputs.

Fine-tuning is useful when you need consistent task behavior or domain-specific formatting. However, it is not automatically the best choice for frequently changing factual knowledge; **Retrieval-Augmented Generation (RAG)** may be more appropriate in such cases.

### Fine-Tuning vs. Instruction Tuning

| Concept | Description | Typical purpose |
|---|---|---|
| **Fine-tuning** | Further training a pretrained model on a target dataset | Adapt the model to a particular task or domain |
| **Supervised fine-tuning (SFT)** | Training on examples containing inputs and desired target responses | Teach specific response patterns and instructions |
| **RAG** | Retrieving relevant external information at inference time | Ground responses in information that can be updated independently of model weights |

### Structure of an SFT Dataset

An SFT training record often contains:

- **Instruction:** The task the model should perform.
- **Optional input/context:** Additional information needed to complete the task.
- **Assistant target:** The desired response the model should learn to produce.

The training objective may calculate loss over all tokens or mask the prompt tokens so that only assistant-response tokens contribute to the loss.

### Chat Templates and Special Tokens

A chat model still processes a **linear sequence of tokens**. A conversation represented as messages with roles such as `system`, `user`, `assistant`, or a tool name must be serialized into the exact format expected by the checkpoint. A tokenizer's **chat template** normally defines the role markers, message boundaries, optional generation prompt, and other control tokens.

A conceptual conversation might be:

```text
system: You are a concise assistant.
user: Explain a causal mask.
assistant: A causal mask blocks attention to future positions.
```

The model generally does **not** receive those labels as plain, self-interpreting metadata. Its trained template converts them into the model-specific token sequence. Different checkpoints can use different tokens and boundary conventions, so do not invent a template or copy one from an unrelated model.

- Use the model's documented template where available: [Hugging Face chat templating guide](https://huggingface.co/docs/transformers/chat_templating).
- Check whether the template already includes beginning/end markers before adding special tokens a second time.
- Distinguish a message ending, an assistant-turn ending, and a tool hand-off when the model's format does so; these meanings are not universally represented by the same token.
- Keep training and inference formatting consistent. A template mismatch can damage instruction-following behavior even if the text content is otherwise identical.

### Important Data and Training Considerations

| Consideration | Why it matters |
|---|---|
| **Dataset deduplication** | Reduces repeated examples and potential data leakage |
| **Train/validation split** | Supports monitoring of generalization during training |
| **Output quality** | Poor target responses can teach undesirable behavior |
| **Privacy controls** | Help prevent sensitive information from being improperly included in training data |
| **Catastrophic forgetting** | Fine-tuning can damage capabilities acquired during pretraining |

### Loss Masking: Training on the Assistant Response

Suppose a token sequence contains both a prompt and its desired answer. If the objective is to train only on the answer, the prompt tokens can be assigned the label `-100`, which PyTorch's cross-entropy loss ignores by default.

For example:

| Token segment | Input tokens | Training labels |
|---|---|---|
| Prompt | `Explain gradient descent` | `-100` for each prompt token |
| Assistant response | `Gradient descent minimizes a loss...` | Corresponding target token IDs |

Conceptually, the objective becomes:

$$
\boxed{\mathcal{L}
=
-\frac{1}{|\mathcal{T}|}
\sum_{t \in \mathcal{T}}
\log P_\theta(y_t \mid y_{<t})}
$$

where:

- $\mathcal{T}$ is the set of token positions included in the loss.
- $y_t$ is the target token at position $t$.
- $P_\theta(y_t \mid y_{<t})$ is the model's predicted probability of that target token given the preceding context.

When prompt tokens are masked, they provide context but do not directly contribute to the loss.

**Practical note:** The following code cell demonstrates the loss-masking mechanism only. It is not a complete model fine-tuning run. Ensure that labels are aligned correctly with the model's shifted prediction targets and that padding tokens are handled appropriately.

### Preference Alignment After Supervised Fine-Tuning

SFT teaches a model to imitate curated target responses. It is often followed by **preference optimization**, which uses judgments about which of two responses is better for the same prompt. Preference alignment can improve helpfulness or instruction following, but its results depend on the preference data, reward design, training method, and evaluation. It does not guarantee factuality or safety.

| Approach | Simplified workflow | Main distinction |
|---|---|---|
| **RLHF (Reinforcement Learning from Human Feedback)** | SFT model → preference comparisons → reward model → optimize the policy using reinforcement learning, commonly with constraints against excessive drift | Uses an explicit learned reward and an RL optimization stage in a common pipeline |
| **DPO (Direct Preference Optimization)** | Train from prompt, preferred answer, and dispreferred answer pairs | Optimizes a preference objective directly relative to a reference model, without needing a separate reward-model training and online RL loop in its standard formulation |
| **RLAIF / AI feedback** | Use model-generated or AI-assisted preferences, often with human quality controls | Scales preference labeling but can inherit evaluator biases and errors |

For DPO, a simplified loss is:

$$
\boxed{\mathcal{L}_{\mathrm{DPO}}(\theta) = -\mathbb{E}\!\left[\log\sigma\!\left(\beta\left(\log\frac{\pi_\theta(y^+\mid x)}{\pi_{\mathrm{ref}}(y^+\mid x)}-\log\frac{\pi_\theta(y^-\mid x)}{\pi_{\mathrm{ref}}(y^-\mid x)}\right)\right)\right]}
$$

Here $x$ is the prompt, $y^+$ is the preferred completion, $y^-$ is the dispreferred completion, $\pi_\theta$ is the policy being trained, $\pi_{\mathrm{ref}}$ is the reference policy, $\sigma$ is the sigmoid function, and $\beta$ controls the preference strength. This formula is included to explain the objective; the notebook does not train a DPO or RLHF model.

References: [InstructGPT / RLHF](https://arxiv.org/abs/2203.02155) · [DPO paper](https://arxiv.org/abs/2305.18290) · [Hugging Face TRL DPO guide](https://huggingface.co/docs/trl/dpo_trainer).

### Choosing Prompting, RAG, Fine-Tuning, or Tools

Choose the lightest approach that reliably solves the actual problem. These approaches are complementary and are often combined.

| Need | A useful starting point | Why / caution |
|---|---|---|
| Change wording, format, or behavior for a clear task | **Prompting** | Fast to iterate; can be brittle across model versions and difficult inputs |
| Use fresh, private, or source-grounded knowledge | **RAG** | Updates the evidence without retraining model weights; retrieval and citation quality must be evaluated |
| Teach a repeated response pattern, style, or domain behavior | **Fine-tuning / SFT** | Can improve consistency; requires high-quality examples and held-out evaluation |
| Perform an action or query an external system | **Tools / function calling** | Gives access to actions or live data; tool permissions and outputs must be validated |
| Improve responses using preference judgments | **Preference optimization** | RLHF/DPO can shape preferences; labels and reward proxies can be biased |

A practical system may combine them—for example, prompting plus RAG plus a restricted search tool. Fine-tuning is not a substitute for current source material, and retrieval is not a guarantee that the model will use evidence correctly. Select methods using task-specific quality, reliability, latency, privacy, and cost measurements.

| Method | Changes model weights? | External knowledge at request time? | Typical reason to use |
|---|---:|---:|---|
| Prompting | No | Only what is in context | Fast iteration, task framing, formatting |
| RAG | Usually no | Yes, retrieved documents | Fresh, private or attributable knowledge |
| Full fine-tuning / SFT | Yes, broadly | Not inherently | Adapt behavior to examples/domain |
| LoRA / QLoRA | Trains adapters; base often frozen | Not inherently | Lower trainable-parameter or memory budget |
| Tool use | Usually no | Yes, via explicit APIs/functions | Calculations, searches, controlled actions |


````python
# Response-only supervised fine-tuning on a tiny character-level instruction dataset.
# The base checkpoint is the tiny GPT trained above; this is a teaching experiment, not useful assistant tuning.
sft_examples = [
    {"prompt": "task attention. answer ", "response": "tokens mix."},
    {"prompt": "task loss. answer ", "response": "loss measures error."},
    {"prompt": "task gradient. answer ", "response": "gradient changes weights."},
    {"prompt": "task model. answer ", "response": "model predicts tokens."},  # held-out example
]
sft_train_examples, sft_val_example = sft_examples[:3], sft_examples[3]
unsupported_chars = sorted({ch for item in sft_examples for ch in item["prompt"] + item["response"] if ch not in char_to_id})
if unsupported_chars:
    raise ValueError(f"SFT strings contain characters absent from the base vocabulary: {unsupported_chars}")

def encode_response_only(record):
    full_ids = torch.tensor([char_to_id[ch] for ch in record["prompt"] + record["response"]],
                            dtype=torch.long, device=DEVICE)
    input_ids = full_ids[:-1].unsqueeze(0)
    labels = full_ids[1:].clone().unsqueeze(0)
    # Labels correspond to the input shifted one position left. Mask prompt targets, retain answer targets.
    prompt_length = len(record["prompt"])
    labels[:, :max(0, prompt_length - 1)] = -100
    return input_ids, labels

sft_model = TinyGPT(VOCAB_SIZE, BLOCK_SIZE, n_embd=64, n_head=4, n_layer=2, dropout=0.05).to(DEVICE)
sft_model.load_state_dict(model.state_dict())
sft_optimizer = torch.optim.AdamW(sft_model.parameters(), lr=1e-3, weight_decay=0.01)

def masked_sft_loss(model_to_eval, record):
    x_sft, y_sft = encode_response_only(record)
    _, loss_sft = model_to_eval(x_sft, y_sft)
    return loss_sft

sft_model.eval()
with torch.no_grad():
    sft_val_before = masked_sft_loss(sft_model, sft_val_example).item()
sft_steps = 45
sft_loss_trace = []
for step in range(sft_steps):
    sft_model.train()
    record = sft_train_examples[step % len(sft_train_examples)]
    x_sft, y_sft = encode_response_only(record)
    _, loss_sft = sft_model(x_sft, y_sft)
    sft_optimizer.zero_grad(set_to_none=True)
    loss_sft.backward()
    torch.nn.utils.clip_grad_norm_(sft_model.parameters(), 1.0)
    sft_optimizer.step()
    sft_loss_trace.append(float(loss_sft.item()))
sft_model.eval()
with torch.no_grad():
    sft_val_after = masked_sft_loss(sft_model, sft_val_example).item()
print("Base model parameter count:", sum(p.numel() for p in model.parameters()))
print("Response-only target characters per training example:", [len(x["response"]) for x in sft_train_examples])
print(f"Held-out response loss before SFT: {sft_val_before:.4f}")
print(f"Held-out response loss after  SFT: {sft_val_after:.4f}")
print(f"Final training response loss: {sft_loss_trace[-1]:.4f}")
plt.figure(figsize=(7, 3)); plt.plot(sft_loss_trace); plt.xlabel("SFT update"); plt.ylabel("Response-only cross-entropy")
plt.title("Tiny GPT supervised fine-tuning (synthetic character-level data)"); plt.grid(alpha=.25); plt.show()

if sft_val_after > sft_val_before:
    print("Validation response loss increased: this tiny training set overfits; lower training loss alone does not establish generalization.")
else:
    print("Validation response loss did not increase on this one held-out example; broader held-out evaluation is still required.")
````

## 17. Parameter-Efficient Fine-Tuning: LoRA & QLoRA

### Why Parameter-Efficient Fine-Tuning?

**Full fine-tuning** updates all model parameters during training. For large language models, this can require substantial GPU memory to store gradients and optimizer states.

**Low-Rank Adaptation (LoRA)** reduces the number of trainable parameters by freezing the pretrained weight matrix and learning a smaller, low-rank update instead.

### 1. Understanding LoRA Mathematically

Let $W$ be a pretrained weight matrix. LoRA represents its adapted version as:

$$
\boxed{W' = W + \Delta W}
$$

where the learned update is:

$$
\boxed{\Delta W = \frac{\alpha}{r}BA}
$$

The matrices have the following dimensions:

$$
\boxed{A \in \mathbb{R}^{r \times d_{\text{in}}}}
$$

$$
\boxed{B \in \mathbb{R}^{d_{\text{out}} \times r}}
$$

Here:

- $d_{\text{in}}$: Input dimension of the weight matrix.
- $d_{\text{out}}$: Output dimension of the weight matrix.
- $r$: Rank of the low-rank update, typically much smaller than the original matrix dimensions.
- $\alpha$: Scaling parameter that controls the magnitude of the update.
- $A$ and $B$: Trainable low-rank matrices.
- $W$: Frozen pretrained weight matrix.

The key idea is that **LoRA learns the update without directly training the original weight matrix**.

### 2. Comparing Trainable Parameter Counts

For a weight matrix of dimensions $d_{\text{out}} \times d_{\text{in}}$, full fine-tuning trains:

$$
\boxed{N_{\text{full}} = d_{\text{out}}d_{\text{in}}}
$$

LoRA trains approximately:

$$
\boxed{N_{\text{LoRA}} = r(d_{\text{in}} + d_{\text{out}})}
$$

For a square matrix with dimension $d$:

$$
\boxed{N_{\text{full}} = d^2}
$$

$$
\boxed{N_{\text{LoRA}} = 2rd}
$$

The reduction in trainable parameters can be expressed as:

$$
\boxed{\text{Reduction (\%)} =
\left(1-\frac{N_{\text{LoRA}}}{N_{\text{full}}}\right)\times 100}
$$

This calculation applies to one adapted matrix and excludes additional parameters, such as biases or other trainable modules, if enabled.

### 3. What Is QLoRA?

**QLoRA (Quantized Low-Rank Adaptation)** combines LoRA adapters with a quantized, frozen base model to reduce memory requirements during fine-tuning.

| Feature | Full fine-tuning | LoRA | QLoRA |
|---|---|---|---|
| Base-model weights | Trainable | Frozen | Quantized and frozen |
| Additional trainable parameters | All model parameters | Low-rank adapters | Low-rank adapters |
| Optimizer states | Typically substantial | Reduced for frozen base weights | Reduced for frozen base weights |
| Base-weight memory | Depends on training precision | Depends on stored precision | Reduced through quantization |
| Main advantage | Full parameter adaptation | Efficient parameter updates | Lower base-model memory requirements |
| Main limitation | High memory requirements | Base model still consumes memory | Quantization and hardware compatibility constraints |

**Important:** QLoRA is not simply LoRA using any lower-precision data type. Standard QLoRA recipes typically use 4-bit quantization with specialized quantization settings. Exact support depends on the hardware, model architecture, and installed library versions. Quantization can introduce accuracy and compatibility trade-offs, so the resulting model should be evaluated on the intended task.


### Quantization Beyond QLoRA

**Quantization** stores numbers using a lower-precision representation. In a simplified symmetric scheme, a value can be approximated as $\hat{x}=s\,q$, where $q$ is a low-bit integer and $s$ is a scale. Real methods differ in how they choose scales, handle outliers, group weights, and calibrate activations.

| Use case | Typical idea | What to validate |
|---|---|---|
| **Post-training quantization (PTQ)** | Quantize an already trained model for inference | Quality loss, supported kernels, and latency on the target hardware |
| **Quantization-aware training (QAT)** | Simulate quantization effects during training or adaptation | Training cost and whether the target inference format is supported |
| **QLoRA** | Keep a quantized base model frozen while training low-rank adapters | Adapter quality, compatibility, memory, and training stability |

Lower-bit weights can reduce the memory footprint, but they do not guarantee faster generation: dequantization overhead, kernels, batch size, and hardware matter. Weight-only quantization also does not necessarily reduce the KV cache, which may become a major memory cost at long context lengths or high concurrency.

````python
class LoRALinear(nn.Module):
    """Educational LoRA wrapper around a frozen Linear layer."""
    def __init__(self, in_features: int, out_features: int, rank: int = 4, alpha: float = 8.0):
        super().__init__()
        self.base = nn.Linear(in_features, out_features, bias=False)
        for p in self.base.parameters(): p.requires_grad = False
        self.A = nn.Parameter(torch.randn(rank, in_features) * 0.02)
        self.B = nn.Parameter(torch.zeros(out_features, rank))
        self.scale = alpha / rank
    def forward(self, x):
        base_out = self.base(x)
        update = (x @ self.A.T) @ self.B.T
        return base_out + self.scale * update

lora_demo = LoRALinear(128, 96, rank=4, alpha=8)
all_params = sum(p.numel() for p in lora_demo.parameters())
trainable_params = sum(p.numel() for p in lora_demo.parameters() if p.requires_grad)
test_input = torch.randn(2, 7, 128)
test_output = lora_demo(test_input)
print("Output shape:", tuple(test_output.shape))
print("Total parameters:", all_params, "trainable parameters:", trainable_params)
print(f"Trainable share: {100*trainable_params/all_params:.2f}%")
````

````python
# Train only LoRA A/B matrices to recover a synthetic low-rank update while keeping the base layer frozen.
torch.manual_seed(SEED + 1)
lora_fit = LoRALinear(in_features=16, out_features=8, rank=2, alpha=4.0).to(DEVICE)
fit_inputs = torch.randn(192, 16, device=DEVICE)
teacher_A = torch.randn(2, 16, device=DEVICE) * 0.20
teacher_B = torch.randn(8, 2, device=DEVICE) * 0.20
with torch.no_grad():
    fit_targets = lora_fit.base(fit_inputs) + lora_fit.scale * ((fit_inputs @ teacher_A.T) @ teacher_B.T)
fit_optimizer = torch.optim.AdamW([lora_fit.A, lora_fit.B], lr=0.03, weight_decay=0.0)
with torch.no_grad():
    fit_initial_loss = F.mse_loss(lora_fit(fit_inputs), fit_targets).item()
fit_losses = []
for _ in range(180):
    fit_optimizer.zero_grad(set_to_none=True)
    fit_loss = F.mse_loss(lora_fit(fit_inputs), fit_targets)
    fit_loss.backward()
    fit_optimizer.step()
    fit_losses.append(float(fit_loss.item()))
with torch.no_grad():
    fit_final_loss = F.mse_loss(lora_fit(fit_inputs), fit_targets).item()
print(f"Synthetic target MSE before adapter training: {fit_initial_loss:.6f}")
print(f"Synthetic target MSE after  adapter training: {fit_final_loss:.6f}")
print("Base weights frozen:", all(not p.requires_grad for p in lora_fit.base.parameters()))
print("Trainable adapter parameter count:", lora_fit.A.numel() + lora_fit.B.numel())
plt.figure(figsize=(7, 3)); plt.plot(fit_losses); plt.xlabel("Adapter update"); plt.ylabel("MSE")
plt.title("LoRA: fitting a synthetic low-rank update"); plt.grid(alpha=.25); plt.show()
````

### 4. Practical Trade-Offs

- **Memory efficiency:** Fewer trainable parameters generally reduce gradient and optimizer-state memory.
- **Adapter portability:** LoRA adapters can often be saved and distributed separately from the base model.
- **Inference requirements:** The base model still needs to be loaded for inference, whether the adapters are applied separately or merged into compatible weights.
- **Rank selection:** A rank that is too small may limit adaptation capacity, while a larger rank reduces the efficiency benefit.
- **Model quality:** LoRA is a parameter-efficient method, but it does not guarantee the same quality as full fine-tuning.

## 18. Retrieval-Augmented Generation (RAG)

### What Is RAG ?

**Retrieval-Augmented Generation (RAG)** combines a language model's learned knowledge, stored in its parameters, with **external retrievable evidence** supplied at inference time. Instead of relying entirely on information encoded during training, a RAG system retrieves relevant documents and provides their content as context for generating an answer.

### The RAG Pipeline

A typical RAG system follows this workflow:

$$
\boxed{
\begin{gathered}
\text{Document Ingestion} \\
\downarrow \\
\text{Chunking} \\
\downarrow \\
\text{Indexing and Representation} \\
\downarrow \\
\text{Query-Based Retrieval} \\
\downarrow \\
\text{Optional Reranking} \\
\downarrow \\
\text{Context Construction} \\
\downarrow \\
\text{Grounded Answer Generation}
\end{gathered}
}
$$

The main stages are:

1. **Document ingestion:** Load source documents into the system.
2. **Chunking:** Divide documents into smaller, retrievable passages.
3. **Indexing:** Compute sparse or dense representations and organize them for retrieval.
4. **Retrieval:** Find chunks relevant to the user's query.
5. **Reranking (optional):** Reorder retrieved chunks using a more targeted relevance model.
6. **Context construction:** Supply the selected evidence to the generator.
7. **Answer generation:** Produce an answer grounded in the retrieved context, ideally with citations to the source material.

### Sparse vs. Dense Retrieval

| Feature | Sparse Lexical Retrieval | Dense Semantic Retrieval |
|---|---|---|
| Representation | Sparse vectors, often based on word or term statistics | Dense vectors produced by an embedding model |
| Typical example | TF–IDF or BM25 | Sentence-transformer embeddings |
| Main strength | Effective lexical matching and term-based relevance | Can retrieve semantically related text with limited word overlap |
| Main limitation | May miss relevant passages expressed using different vocabulary | Quality depends on the embedding model and its domain suitability |

Both approaches can be useful, and production systems may combine them through **hybrid retrieval**.

### Benefits & Limitations

**Benefits**

- **Freshness:** External documents can be updated without retraining the language model.
- **Provenance:** Retrieved sources can support citations and evidence tracing.
- **Domain adaptation:** Private or specialized documents can provide task-specific context.

**Limitations**

- **Retrieval errors:** Relevant documents may not be retrieved.
- **Irrelevant chunks:** Retrieved passages may distract from the actual question.
- **Missing evidence:** The source collection may not contain the required information.
- **Unsupported generation:** A model may still produce claims that the retrieved evidence does not support.

RAG can improve grounding and traceability, but it does **not guarantee factual correctness**.

### Hands-On Demo: Local TF–IDF Retrieval

The demonstration in this notebook uses **TF–IDF vectors and cosine similarity** to retrieve relevant document chunks.

This approach is intentionally lightweight:

- It runs locally without downloading an embedding model.
- It does not require setting up a vector database.
- It returns actual document snippets and relevance scores computed by the code.
- It uses **sparse lexical retrieval**, not dense semantic embeddings.

**Important implementation note:** The final answer construction in this demonstration is extractive and rule-based. It is not an answer generated by an LLM. This distinction lets you inspect retrieval behavior independently of generation quality.

### Key Takeaway

A RAG system has two distinct responsibilities: **retrieve useful evidence and generate a reliable answer from that evidence**. Evaluating retrieval quality separately from answer quality makes it easier to identify whether failures originate in document coverage, ranking, context construction, or generation.


````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

rag_docs = [
    {"id": "D1", "title": "Model training", "text": "The training loop draws token windows, predicts the next token, computes cross entropy, backpropagates gradients, clips gradients and updates parameters."},
    {"id": "D2", "title": "Causal attention", "text": "A causal attention mask prevents each token position from attending to future positions. This prevents leakage during autoregressive training."},
    {"id": "D3", "title": "Retrieval evaluation", "text": "Retrieval quality can be measured with recall at k, precision at k, and mean reciprocal rank using labeled query-document pairs."},
    {"id": "D4", "title": "Deployment", "text": "Inference systems monitor latency, throughput, memory use, failures, and output quality. Quantization can reduce weight memory but may affect quality."},
    {"id": "D5", "title": "Prompt injection", "text": "Retrieved documents are untrusted input. Tool permissions, output validation, source attribution, and human approval reduce system risk."},
]
vectorizer = TfidfVectorizer(stop_words='english')
doc_matrix = vectorizer.fit_transform([d['text'] for d in rag_docs])

def retrieve(query: str, k: int = 3):
    q = vectorizer.transform([query])
    scores = cosine_similarity(q, doc_matrix).ravel()
    order = np.argsort(scores)[::-1][:k]
    return [{**rag_docs[i], 'score': float(scores[i])} for i in order]

query = "How do I measure retrieval quality?"
retrieved = retrieve(query, k=3)
print("Query:", query)
for rank, item in enumerate(retrieved, 1):
    print(f"{rank}. {item['id']} — {item['title']} | cosine={item['score']:.3f}\n   {item['text']}")
````

````python
# Plot actual retrieval scores for this query.
plt.figure(figsize=(7,3.5))
plt.bar([f"{d['id']}\n{d['title']}" for d in retrieved], [d['score'] for d in retrieved])
plt.ylabel('Cosine similarity'); plt.title('TF–IDF retrieval scores for the current query'); plt.ylim(0, max([d['score'] for d in retrieved]+[0.1])*1.15); plt.show()

context = "\n\n".join(f"[{d['id']}] {d['title']}: {d['text']}" for d in retrieved if d['score'] > 0)
rag_prompt = f"Answer the question using only the context. Cite document IDs. Say if the context is insufficient.\n\nContext:\n{context}\n\nQuestion: {query}"
print("Grounded prompt preview:\n", rag_prompt)
````

````python
# Extractive, citation-bearing answer construction with an explicit abstention path.
def extractive_rag_answer(query_text: str, k: int = 3, min_score: float = 1e-8) -> str:
    hits = [item for item in retrieve(query_text, k=k) if item["score"] > min_score]
    if not hits:
        return "Insufficient evidence: no relevant document was retrieved from the local collection."
    cited_lines = [f"[{item['id']}] {item['title']}: {item['text']}" for item in hits]
    return "Evidence retrieved (extractive; no LLM generation):\n" + "\n".join(cited_lines)

print("RAG answer with evidence:\n", extractive_rag_answer("How do I measure retrieval quality?"))
print("\nRAG no-evidence path:\n", extractive_rag_answer("Tell me tomorrow's weather on Mars"))
````

### RAG quality is more than the final answer

- **Recall@k:** among relevant documents, the fraction retrieved in top `k`.
- **Precision@k:** among the top `k` retrieved, the fraction relevant.
- **MRR:** reciprocal rank of the first relevant result, averaged over queries.
- **Faithfulness / support:** whether answer claims are supported by cited context.
- **Citation correctness:** whether a cited document actually entails the claim.

Chunk length, overlap, metadata filtering, hybrid sparse+dense retrieval and reranking can affect recall/precision. Larger chunks may retain context but add distractors; smaller chunks can lose surrounding qualifications. RAG cannot cite documents it never retrieves or has not indexed.


````python
# Tiny retrieval evaluation with synthetic labels, explicitly a made-up teaching benchmark.
eval_cases = [
    ("measure retrieval quality precision recall", {"D3"}),
    ("prevent future token leakage", {"D2"}),
    ("latency and quantization deployment", {"D4"}),
]
def precision_recall_at_k(query_text: str, relevant_ids: set[str], k: int = 3):
    got = retrieve(query_text, k=k)
    predicted = {x['id'] for x in got}
    tp = len(predicted & relevant_ids)
    precision = tp / len(predicted) if predicted else 0.0
    recall = tp / len(relevant_ids) if relevant_ids else 0.0
    return precision, recall, [x['id'] for x in got]

for q, labels in eval_cases:
    p, r, ids = precision_recall_at_k(q, labels, k=3)
    print(f"query={q!r} | top-3={ids} | P@3={p:.2f} R@3={r:.2f} | synthetic relevance labels")
````

### A safe RAG answer pattern

1. Retrieve relevant passages and keep their immutable source identifiers.
2. Put the passages into a clearly delimited *untrusted evidence* section.
3. Instruct the generator to answer only supported claims and cite IDs.
4. Validate that citations exist and, ideally, verify claim-to-passage entailment.
5. Return “insufficient evidence” when support is missing rather than filling gaps with plausible text.

For production, add permissions before retrieval, document deletion/update semantics, embedding model/version tracking, query logging policies, redaction, and testing for cross-user data leakage. Do not rely solely on instructions telling an LLM to ignore a malicious document.

## 19. LLM Agents & Tools Usage

### What Is an LLM Agent?

A chatbot primarily generates responses to user inputs. An **Agentic System** can go further by selecting actions, invoking tools, observing their results, and deciding what to do next. The term *agent* is a broad architectural and product label rather than one fixed algorithm.

### The Agent Execution Loop

A typical tool-using agent follows this cycle:

$$
\boxed{
\begin{gathered}
\text{User Request} \\
\downarrow \\
\text{Reasoning and Action Selection} \\
\downarrow \\
\text{Tool Invocation} \\
\downarrow \\
\text{Observation of Tool Result} \\
\downarrow \\
\text{Next Action or Final Response}
\end{gathered}
}
$$

The agent may repeat the tool-use cycle until it has enough information to complete the task or reaches a stopping condition.

### Common Tools Used by LLM Agents

| Tool | Typical purpose | Example |
|---|---|---|
| **Calculator** | Perform numerical computations | Evaluate an arithmetic expression |
| **Search** | Retrieve relevant information | Find matching passages in a document collection |
| **Database query** | Retrieve structured records | Look up a particular customer's order |
| **External API** | Interact with an external service | Retrieve weather data or create a permitted support ticket |

Tools extend what a model can accomplish, but they also introduce risks. A model-generated action should not automatically be trusted simply because the model selected it.

### Designing Safe and Reliable Tool Use

| Control | Purpose |
|---|---|
| **Narrow tool schemas** | Restrict inputs to explicitly supported fields and types |
| **Allow-listed tools** | Make only approved functions available to the agent |
| **Timeouts and resource limits** | Prevent operations from running indefinitely or consuming excessive resources |
| **Permission checks** | Restrict access to authorized data and actions |
| **Auditable results** | Record tool calls and outcomes for debugging and review |
| **Safe failure paths** | Handle invalid inputs, errors, and unavailable services without unsafe fallback behavior |

**Key principle:** The application should enforce security boundaries independently of the model's instructions. Do not rely on the model alone to decide whether an action is safe or authorized.

- **Calculator:** Evaluates expressions from a restricted arithmetic grammar.
- **Local document search:** Searches the small document collection already defined in the notebook.

The demo intentionally does **not** permit arbitrary shell commands, file writes, or unrestricted network access. This lets you explore how an agent selects tools and uses their results without giving it broad system access.

An LLM agent combines model-generated decisions with external tools and application-controlled execution. Reliable agent design requires not only useful tools, but also strict permissions, input validation, bounded execution, auditable outcomes, and safe handling of failures.


````python
import ast
import operator as op

_ALLOWED_BINOPS = {ast.Add: op.add, ast.Sub: op.sub, ast.Mult: op.mul, ast.Div: op.truediv, ast.Pow: op.pow, ast.Mod: op.mod}
_ALLOWED_UNARYOPS = {ast.UAdd: op.pos, ast.USub: op.neg}

def safe_arithmetic(expression: str, max_abs: float = 1e12):
    """Evaluate numeric arithmetic only; reject names, calls, attributes and oversized values."""
    if len(expression) > 120:
        raise ValueError("Expression too long")
    tree = ast.parse(expression, mode='eval')
    def visit(node):
        if isinstance(node, ast.Expression): return visit(node.body)
        if isinstance(node, ast.Constant) and isinstance(node.value, (int, float)) and not isinstance(node.value, bool):
            value = float(node.value)
        elif isinstance(node, ast.BinOp) and type(node.op) in _ALLOWED_BINOPS:
            left, right = visit(node.left), visit(node.right)
            if isinstance(node.op, ast.Pow) and abs(right) > 10: raise ValueError("Exponent too large")
            value = _ALLOWED_BINOPS[type(node.op)](left, right)
        elif isinstance(node, ast.UnaryOp) and type(node.op) in _ALLOWED_UNARYOPS:
            value = _ALLOWED_UNARYOPS[type(node.op)](visit(node.operand))
        else:
            raise ValueError("Only numeric arithmetic is allowed")
        if not np.isfinite(value) or abs(value) > max_abs:
            raise ValueError("Result is non-finite or outside allowed range")
        return value
    return visit(tree)

def local_doc_search(query: str, k: int = 2):
    return [{"id": d['id'], "title": d['title'], "score": round(d['score'], 3), "text": d['text']}
            for d in retrieve(query, k=k)]

TOOLS = {"calculator": safe_arithmetic, "document_search": local_doc_search}
print("calculator:", TOOLS['calculator']("(17 * 4) / 2"))
print("document search:", TOOLS['document_search']("prompt injection tool permissions", k=2))
try:
    safe_arithmetic("__import__('os').system('echo no')")
except Exception as exc:
    print("Unsafe expression rejected:", type(exc).__name__, str(exc))
````

### Design Controls for Safe LLM Agents

Reliable agent systems require safeguards at both the model and application levels. Because agents can invoke tools and perform multi-step actions, even a small mistake can propagate through repeated tool calls.

#### Essential Safety Controls

| Control | Purpose |
|---|---|
| **Tool allow-list** | Restricts the agent to explicitly approved tools |
| **Argument schema validation** | Ensures tool inputs follow the expected structure and types |
| **Least-privilege credentials** | Gives each tool only the permissions it needs |
| **Request, time, and token budgets** | Limits execution time, resource consumption, and excessive tool calls |
| **Human approval** | Requires confirmation before consequential or irreversible changes |
| **Audit trail** | Records tool invocations, inputs, results, and failures |
| **Safe retry limits** | Prevents endless retries or repeated harmful actions |
| **Instruction–data separation** | Treats untrusted tool output as data rather than trusted instructions |
| **Rollback plans** | Provide a way to reverse changes when technically possible |

#### Why Tool Outputs Must Be Treated Carefully?

Tool outputs can be **incorrect, malicious, outdated, or incomplete**. Retrieved documents and API responses may also contain instructions designed to manipulate the model. A robust agent should validate tool results, preserve the separation between trusted instructions and untrusted content, and avoid treating every returned value as authoritative.

#### Preventing Uncontrolled Agent Loops

An agent loop can amplify a small mistake through repeated actions. For example, an incorrect initial result might cause the agent to invoke additional tools, consume resources, or repeat an unintended operation. A safer execution pattern is:

$$
\boxed{
\begin{gathered}
\text{Validate Request} \\
\downarrow \\
\text{Check Permissions and Budgets} \\
\downarrow \\
\text{Execute Approved Tool} \\
\downarrow \\
\text{Validate Result} \\
\downarrow \\
\text{Continue or Stop Safely}
\end{gathered}
}
$$

Define explicit stopping conditions, enforce retry and resource limits, and test how the system behaves when tools fail or return unexpected results.

**Important security rule:** Never give a demonstration agent unrestricted shell access or destructive permissions. Start with narrowly scoped, allow-listed tools and expand capabilities only when appropriate safeguards are in place.

**Key takeaway:** Agent safety must be enforced by the surrounding application—not merely requested through a prompt. Permission checks, validation, execution limits, human approval, and recovery mechanisms help prevent model errors from becoming consequential system failures.

## 20. Efficient Inference & Deployment

### 1. Understanding Prefill vs. Decode

LLM inference typically consists of two stages: **prefill** and **decode**. Understanding the difference helps explain latency, memory usage, and serving performance.

| Aspect | Prefill | Decode |
|---|---|---|
| **Purpose** | Processes the input prompt | Generates new tokens |
| **Execution** | Processes prompt tokens in parallel | Normally generates one token at a time |
| **Main computation** | Computes initial hidden states and attention keys/values | Computes the next-token prediction using the existing context and KV cache |
| **Performance consideration** | Dense attention computation depends strongly on prompt length | Repeated token generation can be affected by memory bandwidth and cache size |
| **Typical concern** | Time to process long prompts | Time required to generate the response |

The overall inference workflow is:

$$
\boxed{
\text{Input Prompt}
\rightarrow
\text{Prefill}
\rightarrow
\text{KV Cache}
\rightarrow
\text{Autoregressive Decode}
\rightarrow
\text{Generated Response}
}
$$

### 2. Key Techniques for Efficient Inference

**KV cache:** During decoding, the model stores attention keys and values from previously processed tokens. This reduces repeated computation, but the cache consumes additional memory. A simplified relationship is:

$$
\boxed{
\text{KV Cache Memory}
\propto
L \times B \times T_{\text{cache}} \times d_{\text{KV}}
}
$$

where:

- $L$: Number of Transformer layers.
- $B$: Number of concurrent sequences in the batch.
- $T_{\text{cache}}$: Number of cached tokens.
- $d_{\text{KV}}$: Total key/value width per layer.

The precise memory requirement also depends on numerical precision, the number of stored tensors, and the model's attention configuration.

**Batching & Continuous Batching:** Combine multiple requests to improve hardware utilization. Continuous batching allows requests to enter or leave a batch as their generation progresses, but serving systems must balance throughput, latency, and varying sequence lengths.

**Quantization:** Represents weights and, in some approaches, activations using fewer bits. It can reduce memory footprint and memory bandwidth requirements, but quality and kernel compatibility must be evaluated on the target hardware.

### 3. Serving Optimizations Beyond Basic Batching

Several techniques target different bottlenecks. They are not interchangeable, and their benefits depend on the model, workload, software, and hardware.

| Technique | Core idea | Main trade-off |
|---|---|---|
| **Paged KV-cache management** | Allocates KV-cache memory in blocks rather than requiring one contiguous region per sequence | Requires a serving engine that supports the design; scheduling and memory behavior are implementation-dependent |
| **Speculative decoding** | A smaller draft model proposes tokens and a larger target model verifies them, allowing some tokens to be accepted together | Extra draft/verification work can help or hurt depending on acceptance rate and model/hardware balance |
| **Tensor parallelism** | Splits selected tensor operations across devices | Adds communication and can be inefficient at small batch sizes or with slow interconnects |
| **Pipeline parallelism** | Places different model layers on different devices | Can suffer from pipeline bubbles and scheduling complexity |
| **Data parallelism / replicas** | Runs independent model replicas to serve more requests | Increases aggregate capacity but duplicates model memory and requires load balancing |
| **Prefix caching** | Reuses computed prefix state for requests with identical prompt prefixes when supported | Only helps when prefixes repeat and cache validity is managed correctly |

PagedAttention is associated with the vLLM serving design; speculative decoding is a family of methods rather than one universal implementation. Neither provides a guaranteed speed-up for every workload. Measure time to first token, inter-token latency, throughput, peak memory, and tail latency under realistic prompt lengths and concurrency.

References: [PagedAttention paper](https://arxiv.org/abs/2309.06180) · [Speculative decoding paper](https://arxiv.org/abs/2211.17192) · [PyTorch distributed training overview](https://pytorch.org/tutorials/beginner/dist_overview.html).

### Test-Time Compute and Reasoning-Oriented Models

Some systems allocate additional computation at inference time for difficult tasks—for example, by generating more intermediate tokens, sampling multiple candidate solutions, using a verifier or search procedure, or calling tools to check intermediate results. This is often called **test-time compute** or **inference-time scaling**.

It can improve performance on some tasks, but it is not a free improvement and does not guarantee correctness. Additional computation increases latency and cost; candidate generation and verification can also repeat the same mistake. Evaluate task success against a fixed budget and compare quality, latency, token usage, and failure modes. The presence of a long explanation or a reasoning-style answer is not, by itself, evidence that the final answer is correct.

### Attention Variants That Reduce KV-Cache Cost

Standard multi-head attention uses separate key and value heads for each query head. **Multi-Query Attention (MQA)** shares one key/value head across query heads, while **Grouped-Query Attention (GQA)** shares key/value heads across groups of query heads. These designs can reduce KV-cache memory and decoding bandwidth at the cost of architectural and quality trade-offs; the exact benefit depends on the model and implementation. A checkpoint's attention layout must be respected when loading or serving it.

### 4. Throughput vs. Latency

These metrics measure different aspects of serving performance:

| Metric | Meaning | Why it matters |
|---|---|---|
| **Latency** | Time taken for a request or generation step | Affects how quickly a user receives a response |
| **Time to first token (TTFT)** | Time from request submission until the first generated token | Important for interactive applications |
| **Throughput** | Amount of work or number of tokens processed per unit time | Measures overall serving capacity |
| **Peak memory** | Maximum memory consumed during execution | Determines hardware requirements and concurrency limits |

**Important:** Optimizing one metric may hurt another. For example, larger batches can improve throughput while increasing individual request latency.

## 21. Evaluation, Reliability, & Safety

### Why LLM Evaluation Matters?

There is **no single metric that captures overall LLM quality**. Evaluation must match the intended use case and consider multiple dimensions, including task success, factual support, robustness, safety, latency, and cost. A model that performs well on one benchmark may still produce unsupported claims, fail on edge cases, or behave unreliably in real-world applications.

### 1. Common Evaluation Metrics

| Metric / Method | What it measures | Important limitation |
|---|---|---|
| **Perplexity** | How well a language model predicts tokens in an evaluation dataset | Does not directly measure factuality or task success |
| **Exact match / Accuracy** | Whether predictions match expected labels or answers under a defined rule | May penalize valid alternative answers |
| **ROUGE / BLEU** | Different forms of overlap between generated and reference text | Surface-level overlap does not guarantee semantic correctness |
| **Retrieval metrics** | How effectively relevant documents are retrieved and ranked | Good retrieval does not guarantee a grounded final answer |
| **Human evaluation** | Qualities such as usefulness, clarity, and contextual appropriateness | Can be expensive and subject to reviewer variability |

**Key principle:** Select metrics based on the task, and combine automated measurements with human review when necessary.

### 2. Dimensions of Reliability and Safety

LLM evaluation should consider more than average benchmark performance.

| Dimension | What to evaluate |
|---|---|
| **Task success** | Does the model correctly complete the intended task? |
| **Factual support** | Are factual claims supported by reliable evidence? |
| **Robustness** | Does performance remain stable across varied inputs and edge cases? |
| **Safety** | Does the system avoid harmful or prohibited behavior? |
| **Privacy** | Could sensitive information be exposed or improperly reproduced? |
| **Fairness** | Do error rates or outcomes differ across relevant user groups? |
| **Calibration** | Does the model's stated confidence correspond to its actual reliability? |
| **Latency and cost** | Is the system efficient enough for its intended deployment? |
| **Format compliance** | Does the output satisfy the required structure and schema? |

### 3. Common LLM Risks

- **Hallucination:** Producing unsupported or fabricated information.
- **Biased outcomes:** Generating systematically different or unfair responses.
- **Privacy leakage:** Exposing sensitive information.
- **Prompt injection:** Following malicious instructions embedded in user input or retrieved content.
- **Training-data contamination:** Inflating evaluation results through overlap between training data and evaluation examples.
- **Harmful outputs:** Producing unsafe or disallowed content.
- **Excessive refusal:** Rejecting legitimate requests unnecessarily.
- **Brittle formatting:** Failing to follow required output structures.
- **Poor calibration:** Expressing certainty that does not match the available evidence.



### 4. A More Rigorous Evaluation Protocol

A useful evaluation is a repeatable experiment, not just a few impressive examples.

1. **Define the task and acceptance criteria before testing.** Specify allowed sources, required output format, failure costs, and which metric is decision-relevant.
2. **Freeze a held-out evaluation set.** Keep test examples out of prompt iteration, fine-tuning, retrieval-index construction where inappropriate, and threshold selection. Watch for benchmark contamination and near-duplicate examples.
3. **Use task-specific metrics.** For a RAG system, separately measure retrieval recall/ranking, answer correctness, claim support, and citation accuracy. For tool use, measure successful completion, invalid calls, and policy violations—not only answer fluency.
4. **Measure variability.** Report sample size and uncertainty where practical; compare systems on the same examples, and repeat stochastic runs when generation settings make results variable.
5. **Add human review for subjective or high-impact judgments.** Use explicit rubrics and more than one reviewer where feasible. Model-based judges can scale review but may favor certain styles, share biases with the system under test, or be fooled by fluent incorrect answers.
6. **Retest after changes.** Maintain regression, edge-case, and adversarial suites for changes to model weights, prompts, retrieval, tools, templates, and serving configuration.

### 5. Capability, Safety, and Calibration Are Different Questions

- **Capability evaluation:** Can the model perform the task under the specified conditions?
- **Grounding evaluation:** Are claims supported by the evidence available to the system?
- **Safety evaluation:** Does the complete application handle adversarial inputs and high-risk cases appropriately?
- **Calibration:** Do confidence estimates or abstention decisions match observed error rates? Fluent language is not a calibrated probability.

A system can score well on a benchmark and still fail in deployment because its users, documents, tools, or input distribution differ from the evaluation setup. Treat evaluations as evidence with a stated scope, not as universal guarantees.

Useful reference: [Stanford HELM](https://github.com/stanford-crfm/helm) is an example of a holistic, transparent model-evaluation framework; inspect its current status and benchmark versions before adopting it in a new project.

Test using realistic examples and adversarial cases. Stratify results by relevant subgroups and failure categories, measure regressions, and establish human escalation procedures for high-impact decisions.

### 6. Claim-Support & Safety Checks for RAG

For a Retrieval-Augmented Generation (RAG) system, evaluate each factual claim against its cited source passage — not merely whether the answer sounds plausible. A useful review process is:

$$
\boxed{
\begin{gathered}
\text{Generated Claim} \\
\downarrow \\
\text{Locate Supporting Passage} \\
\downarrow \\
\text{Check Evidence Relevance} \\
\downarrow \\
\text{Verify Claim Is Entailed} \\
\downarrow \\
\text{Accept, Revise, or Abstain}
\end{gathered}
}
$$

Consider the following checks:

- **Evidence relevance:** Does the cited passage address the claim?
- **Evidence sufficiency:** Does the passage contain enough information to support it?
- **Faithfulness:** Does the answer stay within what the evidence supports?
- **Citation accuracy:** Does the citation point to the passage that supports the claim?
- **Abstention:** Does the system decline to make a claim when sufficient evidence is unavailable?

In high-stakes settings, define acceptable evidence and explicit abstention criteria in advance. Where consequences warrant it, human review should be independent of the model's own self-assessment.

### Key Takeaway

Reliable LLM deployment requires **continuous, use-case-specific evaluation**, not reliance on a single benchmark or the model's own confidence. Measure task performance, verify factual support, test safety and robustness, document remaining risks, and monitor the system after release.


````python
# Runnable text-metric example: exact match and token overlap F1 on reference/candidate pairs.
metric_examples = [
    {"reference": "the model predicts the next token", "prediction": "the model predicts the next token"},
    {"reference": "retrieve relevant documents and cite evidence", "prediction": "retrieve documents and cite evidence"},
    {"reference": "the system abstains when evidence is missing", "prediction": "the system answers when evidence is missing"},
]

def normalized_exact_match(reference: str, prediction: str) -> bool:
    return " ".join(reference.lower().split()) == " ".join(prediction.lower().split())

def token_overlap_scores(reference: str, prediction: str):
    ref_tokens, pred_tokens = reference.lower().split(), prediction.lower().split()
    ref_counts, pred_counts = Counter(ref_tokens), Counter(pred_tokens)
    overlap = sum((ref_counts & pred_counts).values())
    precision = overlap / len(pred_tokens) if pred_tokens else 0.0
    recall = overlap / len(ref_tokens) if ref_tokens else 0.0
    f1 = (2 * precision * recall / (precision + recall)) if precision + recall else 0.0
    return precision, recall, f1

exact_values, f1_values = [], []
for index, example in enumerate(metric_examples, 1):
    exact = normalized_exact_match(example["reference"], example["prediction"])
    precision, recall, f1 = token_overlap_scores(example["reference"], example["prediction"])
    exact_values.append(float(exact)); f1_values.append(f1)
    print(f"case {index}: exact_match={exact} | token_precision={precision:.2f} | token_recall={recall:.2f} | token_F1={f1:.2f}")
print(f"Mean exact match: {np.mean(exact_values):.3f}")
print(f"Mean token-overlap F1: {np.mean(f1_values):.3f}")
print("These simple overlap metrics are useful diagnostics, not factuality or safety guarantees.")
````

## 22. Beyond Standard Transformers: Transformers vs. Mamba

### 22.1 Understanding the Architecture Landscape

Not every language model processes sequences in the same way. Different architectures use different mechanisms to represent context, process sequences, and allocate computation. A useful way to organize the landscape is:

$$
\boxed{
\begin{gathered}
\text{Sequence Modeling Architectures} \\[4pt]
\downarrow \\[4pt]
\begin{aligned}
&\text{Transformer-Based Models} \\
&\text{State-Space Models (SSMs)} \\
&\text{Hybrid Architectures}
\end{aligned}
\end{gathered}
}
$$

Other design choices, such as **Mixture of Experts (MoE)**, can be combined with these architectural families rather than treated as mutually exclusive alternatives.

### 22.2 Major Architecture Families

| Architecture | How it processes information | Typical applications | Main consideration |
|---|---|---|---|
| **BERT-like encoder-only** | Uses bidirectional context to build representations | Classification, embeddings, token labeling | Not designed for native left-to-right generation |
| **T5-like encoder–decoder** | Encodes an input sequence and generates an output sequence | Translation, summarization, conditional generation | Uses separate encoder and decoder components |
| **GPT-like decoder-only** | Predicts the next token using preceding tokens | Chat, code generation, text continuation | Standard dense self-attention has quadratic sequence-length scaling |
| **Mixture of Experts (MoE)** | Routes each token through a selected subset of expert networks | Scaling model capacity while limiting activated parameters per token | Routing, communication, and load balancing add complexity |
| **State-space models (SSMs)** | Processes sequences through a structured evolving hidden state | Long-sequence modeling and streaming-oriented workloads | Performance depends on the state-space formulation and implementation |
| **Mamba** | Uses input-dependent selective state-space dynamics | Sequence modeling with linear sequence-length scaling in its selective scan formulation | Actual speed and quality depend on hardware, kernels, model size, and workload |
| **Hybrid models** | Combine attention, state-space layers, or other components | Applications requiring different computational trade-offs | More architectural and implementation complexity |

**Important distinction:** BERT, T5, and GPT describe different Transformer architecture patterns. Mamba is a selective state-space architecture. MoE describes a conditional-computation strategy that can be incorporated into a larger model architecture.


### 22.3 How Transformers and Mamba Process Sequences!!

#### A. Transformer: Content-Based Attention

Self-attention allows each token to compare its representation with other tokens in the permitted context and aggregate relevant information. Conceptually:

$$
\boxed{
\text{Token Representations}
\rightarrow
\text{Attention Scores}
\rightarrow
\text{Weighted Information Mixing}
}
$$

Each token can directly interact with other permitted positions through the attention operation.

For standard dense self-attention, the number of pairwise token interactions scales quadratically with sequence length:

$$
\boxed{\text{Attention Scaling} \sim O(T^2)}
$$

where $T$ is the sequence length.

This interaction pattern supports flexible context-dependent retrieval, but long sequences can require substantial computation and memory.

#### B. Mamba: Selective State-Space Processing

Mamba belongs to the **State-Space Model (SSM)** family. It processes a sequence through an evolving internal state, using input-dependent parameters that allow it to selectively preserve or discard information. Conceptually:

$$
\boxed{
\text{Input Token}
\rightarrow
\text{Update Hidden State}
\rightarrow
\text{Produce Output}
}
$$

The state evolves as the model processes the sequence, rather than explicitly constructing a dense pairwise attention matrix at every layer. The original Mamba paper introduced a selective scan formulation with linear scaling in sequence length:

$$
\boxed{\text{Selective Scan Scaling} \sim O(T)}
$$

This refers to sequence-length scaling for the relevant computation, not a guarantee of lower wall-clock latency in every setting.

### 22.4 Transformer vs. Mamba: Side-by-Side Comparison

| Dimension | Transformer | Mamba / Selective SSM |
|---|---|---|
| **Core mechanism** | Attention between token representations | Input-dependent state updates |
| **Information access** | Direct interactions between permitted context positions | Information is carried through the evolving state |
| **Sequence-length scaling** | Standard dense self-attention has quadratic pairwise interactions | Selective scan has linear sequence-length scaling |
| **Long-context considerations** | Attention computation and cache requirements can become expensive | Can offer advantages for long sequences, depending on state size and implementation |
| **Training parallelism** | Highly parallelizable across sequence positions | Efficient parallel scan algorithms support parallel computation |
| **Inference behavior** | Autoregressive generation typically maintains a KV cache | State-space inference maintains and updates a recurrent state |
| **Engineering ecosystem** | Mature tools, optimized kernels, and broad deployment support | Performance depends on implementation, kernel support, and hardware |
| **Best choice** | Depends on the task and deployment constraints | Depends on the task and deployment constraints |

### 22.5 Why Linear Scaling Does Not Automatically Mean Faster?

A lower asymptotic complexity is useful, but it does not tell the whole performance story. Actual performance depends on:

- **Sequence length:** Short and long sequences may favor different implementations.
- **Hardware:** GPU architecture, memory bandwidth, and compute capacity matter.
- **Kernel quality:** Optimized implementations can substantially change performance.
- **Batch size and concurrency:** Serving workload affects resource utilization.
- **Model dimensions:** Hidden-state size, model width, and number of layers influence computation.
- **Target task:** Quality on language generation, retrieval, reasoning, or other tasks must be evaluated independently.

Transformers have excellent parallel training support and a mature ecosystem. State-space models may be attractive when long-sequence processing or memory and throughput constraints are important. Neither architecture is universally faster or more accurate.

### 22.6 Where MoE & Hybrid Architectures Fit?

**Mixture of Experts (MoE)** introduces multiple expert networks and uses a routing mechanism to select which experts process each token. Unlike a dense model, an MoE model can have a large total parameter count while activating only a subset of its expert parameters for a given token. A simplified processing path is:

$$
\boxed{
\text{Token}
\rightarrow
\text{Router}
\rightarrow
\text{Selected Experts}
\rightarrow
\text{Combined Output}
}
$$

This can increase model capacity without activating every expert per token. However, routing overhead, communication costs, and uneven expert utilization can reduce the practical efficiency gains.

**Hybrid architectures** combine mechanisms such as attention and state-space layers to balance information access, sequence scaling, and computational cost. These are complementary design choices: a model can use a decoder-only backbone, MoE layers, state-space components, or combinations of them.

### 22.7 How to Choose an Architecture?

| Your priority | What to investigate |
|---|---|
| Broad compatibility and mature tooling | Transformer-based models |
| Direct content-dependent interactions across context | Attention-based architectures |
| Long-sequence processing under memory or throughput constraints | State-space models such as Mamba |
| High parameter capacity with conditional computation | MoE architectures |
| A combination of attention and efficient sequence processing | Hybrid architectures |

Treat these as starting points rather than universal recommendations. Evaluate candidate models using the actual sequence lengths, hardware, latency targets, memory budget, and quality requirements of your application.

**Key takeaway:** Transformers and Mamba use different mechanisms for sequence modeling. Transformers provide flexible attention-based interactions, while Mamba uses selective state-space processing with linear sequence-length scaling in its selective scan formulation. The best architecture depends on measured performance, task quality, and deployment constraints — not complexity notation alone.


### 22.8 Implementing and Benchmarking a Selective State-Space Model

The following code implements a **small educational input-dependent state-space reference layer** so its state update and tensor shapes can be inspected. It is **not a complete or optimized Mamba implementation**: it does not reproduce the full published architecture or fused selective-scan kernels. Its Python time-step loop is for conceptual clarity, not a production-speed comparison.

The timing cell compares two small reference implementations under the same simple conditions. Treat its results as a local demonstration only. A fair architecture comparison requires comparable model quality, optimized implementations, warm-up, appropriate device synchronization, multiple repetitions, realistic sequence lengths and batches, and evaluation of both task quality and memory use.

````python
class SelectiveSSMReference(nn.Module):
    """Educational input-dependent diagonal state-space layer; not a complete production Mamba implementation."""
    def __init__(self, d_model: int = 64, d_state: int = 32):
        super().__init__()
        self.d_state = d_state
        self.input_gate = nn.Linear(d_model, 2 * d_model)
        self.selective_params = nn.Linear(d_model, 3 * d_state)
        self.state_to_model = nn.Linear(d_state, d_model)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        batch, seq_len, _ = x.shape
        u, gate = self.input_gate(x).chunk(2, dim=-1)
        delta, b_in, c_out = self.selective_params(u).chunk(3, dim=-1)
        delta = F.softplus(delta)  # positive, input-dependent update rates
        state = torch.zeros(batch, self.d_state, dtype=x.dtype, device=x.device)
        outputs = []
        for t in range(seq_len):
            decay = torch.exp(-delta[:, t, :])
            state = decay * state + (1.0 - decay) * torch.tanh(b_in[:, t, :])
            outputs.append(torch.tanh(c_out[:, t, :]) * state)
        state_sequence = torch.stack(outputs, dim=1)
        return x + self.state_to_model(state_sequence) * torch.sigmoid(gate)


torch.manual_seed(SEED)
ssm_block = SelectiveSSMReference(d_model=64, d_state=32).to(DEVICE)
ssm_input = torch.randn(2, 24, 64, device=DEVICE, requires_grad=True)
ssm_output = ssm_block(ssm_input)
ssm_output.square().mean().backward()
print("Selective SSM input/output shapes:", tuple(ssm_input.shape), tuple(ssm_output.shape))
print("Finite input gradients:", bool(torch.isfinite(ssm_input.grad).all().item()))
print("Selective SSM parameter count:", sum(p.numel() for p in ssm_block.parameters()))

# Compare illustrative implementations across sequence lengths. Timing is not a general architecture ranking.
def _median_forward_ms(module, inputs, repeats=3):
    module.eval()
    with torch.no_grad():
        _ = module(inputs)
        _sync_device()
        samples = []
        for _ in range(repeats):
            _sync_device(); start = time.perf_counter()
            _ = module(inputs)
            _sync_device(); samples.append((time.perf_counter() - start) * 1000)
    return float(np.median(samples))

attention_reference = CausalSelfAttention(n_embd=64, n_head=4, block_size=64, dropout=0.0).to(DEVICE).eval()
ssm_reference = SelectiveSSMReference(d_model=64, d_state=32).to(DEVICE).eval()
print("\nToy forward-pass timing (same batch/width; not a production Mamba benchmark):")
for sequence_length in [16, 32, 64]:
    sample = torch.randn(1, sequence_length, 64, device=DEVICE)
    attention_ms = _median_forward_ms(attention_reference, sample)
    ssm_ms = _median_forward_ms(ssm_reference, sample)
    print(f"T={sequence_length:2d} | dense attention={attention_ms:8.3f} ms | selective SSM reference={ssm_ms:8.3f} ms")
````

### 22.9 What to compare in a fair experiment?

| Dimension | Transformer | SSM / Mamba-style | What a fair benchmark must control |
|---|---|---|---|
| Sequence scaling | Standard dense attention has O(T²) pair interactions | Scan-based sequence processing can scale O(T) in T | Same sequence lengths, batch, dtype and kernels |
| Training parallelism | Strong parallelism across tokens in attention | Selective scan kernels parallelize recurrence; implementation matters | Warm-up, compilation and device synchronization |
| Long-context memory | KV cache grows with cached context during autoregressive decoding | Recurrent state can have a bounded per-layer state independent of full token history | State, cache, activation and batch memory included |
| Content-based interaction | Direct token-to-token attention | Input-dependent state updates; interaction mechanism differs | Task quality, not only throughput |
| Ecosystem | Broad mature tooling and pretrained checkpoints | Ecosystem and checkpoint availability vary by implementation | Equivalent model quality/training data and software maturity |

## 23. Multimodal LLMs & Vision-Language Models

### 23.1 What Are Multimodal LLMs?

A **Multimodal Large Language Model (LLM)** processes information from more than one modality, such as text, images, audio, or video. Unlike a text-only language model, which operates primarily on tokenized text, a multimodal model incorporates representations from different types of input and combines them with language-model representations to perform tasks that require cross-modal understanding.

### 23.2 How Does a Multimodal Model Work?

A vision-language model (VLM), for example, may use an image encoder to convert image pixels or patches into vector representations. These representations are then aligned with or integrated into the language model's processing pipeline. A simplified architecture is:

$$
\boxed{
\begin{gathered}
\text{Image Input} \\
\downarrow \\
\text{Image Encoder} \\
\downarrow \\
\text{Visual Representations} \\
\downarrow \\
\text{Alignment or Fusion with Text} \\
\downarrow \\
\text{Language Model} \\
\downarrow \\
\text{Text Response}
\end{gathered}
}
$$

The exact architecture varies. Some models project visual features into the language model's embedding space, while others use cross-attention or more complex fusion mechanisms.

**Key idea:** Different modalities require different encoding strategies, but their representations must interact in a way that supports the target task.

### 23.3 How Different Modalities Are Represented?

| Modality | Typical representation approach | Example input |
|---|---|---|
| **Text** | Token IDs and learned token embeddings | A question or paragraph |
| **Images** | Pixel patches or visual features produced by an image encoder | A photograph, chart, or scanned document |
| **Audio** | Waveform or spectrogram encoders, or learned audio tokens | Speech, music, or environmental sounds |
| **Video** | Sampled frames combined with temporal information | A recorded event or video clip |

These are families of design approaches, not a single universal architecture. A model may use separate encoders, a shared representation space, cross-attention, token-based interfaces, or combinations of these techniques.

### 23.4 Common Multimodal Applications

| Task | What the model does | Example |
|---|---|---|
| **Image captioning** | Generates a textual description of an image | Describing a photograph |
| **Visual question answering (VQA)** | Answers questions about visual content | Identifying an object in an image |
| **OCR and document understanding** | Extracts and interprets text and structure in documents | Summarizing a scanned invoice |
| **Image-text retrieval** | Finds images relevant to a text query, or vice versa | Searching for images using a natural-language description |
| **Speech recognition** | Converts spoken audio into text | Transcribing a lecture |
| **Text-to-speech (TTS)** | Generates spoken audio from text | Producing an audible assistant response |
| **Video understanding** | Interprets frames and temporal relationships | Summarizing the events in a video |

### 23.5 Important Limitations

Multimodal capabilities do not guarantee reliable interpretation. Models can fail when visual, acoustic, or temporal information is ambiguous or difficult to process.

| Limitation | Why it matters |
|---|---|
| **Small text in images** | Fine print and low-resolution text may be misread |
| **Chart and table interpretation** | Numerical values, axes, legends, and relationships can be misunderstood |
| **Spatial reasoning errors** | The model may confuse object positions, sizes, or relationships |
| **Temporal reasoning errors** | The model may miss event order, motion, or changes across video frames |
| **Modality mismatch** | Visual or audio features may not align well with the language model's representations |
| **Hallucinated details** | The model may describe objects, events, or text that are not actually present |
| **Resolution and sampling limits** | Important details may be lost when images are resized or video frames are sampled sparsely |

### Key Takeaway

Multimodal LLMs extend language-model capabilities by combining text with visual, audio, or video information. Their architectures differ in how they encode and fuse these representations, and their reliability depends on the modality, task, input quality, and evaluation method. Always verify outputs against the original input when accuracy matters.


````python
# Self-contained, synthetic vision example: create an image and turn it into patch vectors.
from PIL import Image, ImageDraw
image_size = 96
synthetic_image = Image.new('RGB', (image_size, image_size), 'white')
draw = ImageDraw.Draw(synthetic_image)
draw.rectangle((8, 8, 42, 42), fill=(30, 120, 200))
draw.ellipse((50, 15, 86, 51), fill=(230, 120, 40))
draw.line((8, 75, 88, 75), fill=(20, 150, 90), width=6)
plt.figure(figsize=(3.5,3.5)); plt.imshow(synthetic_image); plt.axis('off'); plt.title('Synthetic teaching image (not external evidence)'); plt.show()

image_array = np.asarray(synthetic_image, dtype=np.float32) / 255.0
PATCH = 16
patches = []
for y in range(0, image_size, PATCH):
    for x in range(0, image_size, PATCH):
        patches.append(image_array[y:y+PATCH, x:x+PATCH, :].reshape(-1))
patch_tensor = torch.tensor(np.stack(patches), dtype=torch.float32)
patch_projector = nn.Linear(PATCH*PATCH*3, 32)
patch_embeddings = patch_projector(patch_tensor)
print("Image array:", image_array.shape)
print("Patches:", tuple(patch_tensor.shape), "meaning [number_of_patches, flattened_RGB_patch]")
print("Patch embeddings:", tuple(patch_embeddings.shape), "meaning [number_of_patches, embedding_dim]")
````

````python
# Small paired image/text alignment experiment on synthetic shapes (CLIP-like contrastive objective).
# The dataset is intentionally tiny and synthetic; success demonstrates the mechanism, not real-world vision understanding.
from PIL import Image, ImageDraw

def make_synthetic_pair(kind: str, size: int = 32):
    image = Image.new("RGB", (size, size), "white")
    painter = ImageDraw.Draw(image)
    if kind == "blue square":
        painter.rectangle((7, 7, 24, 24), fill=(20, 100, 220))
    elif kind == "orange circle":
        painter.ellipse((7, 7, 24, 24), fill=(240, 130, 20))
    elif kind == "green line":
        painter.line((4, 16, 28, 16), fill=(20, 170, 70), width=5)
    else:
        raise ValueError(kind)
    return image

paired_captions = ["blue square", "orange circle", "green line"]
paired_images_pil = [make_synthetic_pair(caption) for caption in paired_captions]
paired_image_batch = torch.tensor(
    np.stack([np.asarray(im, dtype=np.float32).transpose(2, 0, 1) / 255.0 for im in paired_images_pil]),
    dtype=torch.float32, device=DEVICE)
text_vocabulary = sorted({word for caption in paired_captions for word in caption.split()})
text_word_to_id = {word: index for index, word in enumerate(text_vocabulary)}
paired_text_ids = torch.tensor([[text_word_to_id[w] for w in caption.split()] for caption in paired_captions],
                               dtype=torch.long, device=DEVICE)

class TinyVisualEncoder(nn.Module):
    def __init__(self, out_dim: int = 16):
        super().__init__()
        self.net = nn.Sequential(nn.Flatten(), nn.Linear(3 * 32 * 32, 64), nn.GELU(), nn.Linear(64, out_dim))
    def forward(self, image_tensor):
        return F.normalize(self.net(image_tensor), dim=-1)

class TinyTextEncoder(nn.Module):
    def __init__(self, vocab_size: int, out_dim: int = 16):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, out_dim)
        self.projection = nn.Linear(out_dim, out_dim, bias=False)
    def forward(self, token_ids):
        pooled = self.embedding(token_ids).mean(dim=1)
        return F.normalize(self.projection(pooled), dim=-1)

image_encoder_demo = TinyVisualEncoder().to(DEVICE)
text_encoder_demo = TinyTextEncoder(len(text_vocabulary)).to(DEVICE)
alignment_optimizer = torch.optim.AdamW(
    list(image_encoder_demo.parameters()) + list(text_encoder_demo.parameters()), lr=0.01, weight_decay=0.001)
pair_labels = torch.arange(len(paired_captions), device=DEVICE)

def paired_contrastive_loss():
    img_vecs = image_encoder_demo(paired_image_batch)
    txt_vecs = text_encoder_demo(paired_text_ids)
    similarities = img_vecs @ txt_vecs.T / 0.1
    loss_i = F.cross_entropy(similarities, pair_labels)
    loss_t = F.cross_entropy(similarities.T, pair_labels)
    return (loss_i + loss_t) / 2, similarities

with torch.no_grad():
    align_loss_before, _ = paired_contrastive_loss()
alignment_losses = []
for _ in range(180):
    alignment_optimizer.zero_grad(set_to_none=True)
    align_loss, _ = paired_contrastive_loss()
    align_loss.backward()
    alignment_optimizer.step()
    alignment_losses.append(float(align_loss.item()))
with torch.no_grad():
    align_loss_after, similarity_matrix = paired_contrastive_loss()
print("Image batch shape:", tuple(paired_image_batch.shape), "| text token shape:", tuple(paired_text_ids.shape))
print(f"Image/text contrastive loss before training: {align_loss_before.item():.4f}")
print(f"Image/text contrastive loss after  training: {align_loss_after.item():.4f}")
print("Most similar caption for each synthetic image:")
for row, match in enumerate(similarity_matrix.argmax(dim=1).tolist()):
    print(f"  image {row + 1} → {paired_captions[match]!r} (paired caption: {paired_captions[row]!r})")
plt.figure(figsize=(6, 3.8)); plt.plot(alignment_losses); plt.xlabel("Training step"); plt.ylabel("Contrastive loss")
plt.title("Synthetic image–text alignment experiment"); plt.grid(alpha=.25); plt.show()
````
